```python
# ╔══════════════════════════════════════════════════════════════════════════════╗
# ║  FLOW LOOP REPORT GENERATOR — FULL INTERACTIVE UI EDITION                   ║
# ╚══════════════════════════════════════════════════════════════════════════════╝

import subprocess, sys
def install(pkg):
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])
for p in ["google-genai", "reportlab", "pandas", "numpy", "scipy",
          "matplotlib", "openpyxl", "ipywidgets", "Pillow",
          "openai", "anthropic", "groq", "requests"]:
    try: __import__(p.replace("-", "_").split("==")[0])
    except ImportError: install(p)

import os, re, io, json, math, warnings, textwrap, copy, base64, traceback
from pathlib import Path
from datetime import datetime
from dataclasses import dataclass, field
from typing import Optional
import numpy as np
import pandas as pd
from scipy import signal, stats
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.gridspec import GridSpec
from reportlab.lib.pagesizes import letter, A4
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import inch, mm
from reportlab.lib.enums import TA_LEFT, TA_CENTER, TA_RIGHT, TA_JUSTIFY
from reportlab.platypus import (BaseDocTemplate, PageTemplate, Frame, NextPageTemplate,
    Paragraph, Spacer, Image as RLImage, Table, TableStyle, PageBreak,
    KeepTogether, HRFlowable)
from reportlab.pdfgen import canvas as pdfcanvas
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output, Image as IPImage
from PIL import Image as PILImg
warnings.filterwarnings('ignore')

# ══════════════════════════════════════════════════════════════════
#  HELPER: Handle file upload across ipywidgets versions
# ══════════════════════════════════════════════════════════════════

def get_uploaded_file(upload_widget):
    if not upload_widget.value:
        return None, None
    val = upload_widget.value
    if isinstance(val, tuple):
        if len(val) >= 2:
            return val[0], val[1]
    if isinstance(val, list) and len(val) > 0:
        first = val[0]
        if isinstance(first, tuple) and len(first) >= 2:
            return first[0], first[1]
        if isinstance(first, dict):
            for name, info in first.items():
                if isinstance(info, dict) and 'content' in info:
                    return name, info['content']
                elif isinstance(info, bytes):
                    return name, info
    if isinstance(val, dict):
        for name, info in val.items():
            if isinstance(info, dict) and 'content' in info:
                return name, info['content']
            elif isinstance(info, bytes):
                return name, info
            elif isinstance(info, tuple) and len(info) >= 2:
                return info[0], info[1]
    return None, None

# ══════════════════════════════════════════════════════════════════
#  CONFIG
# ══════════════════════════════════════════════════════════════════

@dataclass
class ReportConfig:
    api_key: str = ""
    model_name: str = "gemini-2.0-flash"
    ai_provider: str = "Gemini"
    project_title: str = "FLOW LOOP COMMISSIONING"
    project_subtitle: str = "Effect of Flowrate Changes on Hydraulic Transients"
    project_sub2: str = "Step-Up & Step-Down Analysis  -  With Orifice Plate"
    plan_id: str = "Plan N01 - Trial N03"
    header_left: str = "Flow Loop Commissioning - Flowrate Change Effects"
    header_right: str = "Plan N01 - Trial N03"
    footer_right: str = "With Orifice Plate | Step Transition Study"
    flow_conditions: list = field(default_factory=lambda: [200, 300, 400, 500])
    baseline_flow: float = 100.0
    sampling_rate: float = 10.0
    orifice_installed: bool = True
    cycles_per_condition: str = "3-5 cycles"
    pump_driver: str = "Variable Frequency Drive (VFD)"
    transition_sigma: float = 3.0
    pre_transition_window: float = 5.0
    post_transition_window: float = 15.0
    settling_sigma: float = 2.0
    ramp_pct: float = 0.95
    psd_nperseg: int = 256
    page_width: float = 8.5 * inch
    page_height: float = 11.0 * inch
    margin_left: float = 45
    margin_right: float = 45
    margin_top: float = 60
    margin_bottom: float = 50
    header_y: float = 10.5 * inch
    footer_y: float = 30
    output_dir: str = "report_output"
    output_filename: str = "flowrate_effects_report.pdf"
    color_primary: str = "#1f4e78"
    color_secondary: str = "#4472c4"
    color_text: str = "#2b2b2b"
    color_header_bg: str = "#d6e4f0"
    color_light_gray: str = "#f2f2f2"
    condition_colors: list = field(default_factory=lambda: ["#4472C4", "#70AD47", "#ED7D31", "#FF0000"])
    pressure_color: str = "#FF0000"

    def __post_init__(self):
        if not self.api_key: self.api_key = os.environ.get("GOOGLE_API_KEY", "")
        Path(self.output_dir).mkdir(parents=True, exist_ok=True)

# ══════════════════════════════════════════════════════════════════
#  DATA LOADER
# ══════════════════════════════════════════════════════════════════

class FlowLoopDataLoader:
    CALIBRATIONS = {}
    FLOW_CHANNEL = None
    PRESSURE_CHANNEL = None

    def __init__(self, config, excel_path=None, raw_bytes=None):
        self.cfg = config; self.excel_path = excel_path; self.raw_bytes = raw_bytes
        self.is_real_data = False; self.data_units = {'flowrate': 'unknown', 'pressure': 'unknown'}
        np.random.seed(42)

    def load(self):
        if self.raw_bytes is not None:
            try:
                print("  Loading REAL data from uploaded bytes...")
                data = self._load_from_bytes(); self.is_real_data = True
                print("  Loaded real data for", len(data), "condition(s)"); return data
            except Exception as e: print("  Bytes load failed:", e, "- Using synthetic fallback.")
        elif self.excel_path and Path(self.excel_path).exists():
            try:
                print("  Loading REAL data from:", self.excel_path)
                data = self._load_from_file(); self.is_real_data = True
                print("  Loaded real data for", len(data), "condition(s)"); return data
            except Exception as e: print("  File load failed:", e, "- Using synthetic fallback.")
        print("  Using SYNTHETIC demo data")
        return FlowLoopDataGenerator(self.cfg).generate_all_conditions()

    def _load_from_file(self): return self._parse_excel(pd.ExcelFile(self.excel_path))
    def _load_from_bytes(self): return self._parse_excel(pd.ExcelFile(io.BytesIO(self.raw_bytes)))

    def _parse_excel(self, xls):
        all_data = {}
        for target in self.cfg.flow_conditions:
            sheet_name = "100_{}".format(target)
            if sheet_name in xls.sheet_names:
                df = pd.read_excel(xls, sheet_name=sheet_name)
                all_data[sheet_name] = self._process_condition(df, target); print("    Loaded sheet", sheet_name)
        if all_data: return all_data
        df = pd.read_excel(xls, sheet_name=0)
        if 'time' in df.columns and 'flowrate' in df.columns and 'pressure' in df.columns:
            print("    Single-sheet with standard columns detected...")
            self.data_units = {'flowrate': 'gpm', 'pressure': 'psi'}; return self._auto_detect_conditions(df)
        print("    Trying horizontal channel format..."); return self._parse_horizontal_format(xls)

    def _parse_horizontal_format(self, xls):
        df_raw = pd.read_excel(xls, sheet_name=0, header=None); channel_row_idx = None
        for idx in range(min(30, len(df_raw))):
            row_vals = [str(v).strip() for v in df_raw.iloc[idx].values if pd.notna(v)]
            if 'Channel name' in row_vals: channel_row_idx = idx; break
        if channel_row_idx is None: raise ValueError("Cannot find 'Channel name' row")
        ch_names = []
        for val in df_raw.iloc[channel_row_idx].values:
            v = str(val).strip()
            if v not in ('Channel name', 'nan', '', 'None') and pd.notna(val): ch_names.append(v)
        if not ch_names: raise ValueError("No channel names found")
        print("    Found channels:", ch_names)
        data_start = channel_row_idx + 2; n_cols = df_raw.shape[1]; n_channels = len(ch_names)
        cols_per_channel = max(1, n_cols // n_channels); time_values = []; channel_data = {ch: [] for ch in ch_names}
        for row_idx in range(data_start, len(df_raw)):
            row = df_raw.iloc[row_idx].values
            try: t = float(row[0])
            except (ValueError, TypeError, IndexError): continue
            time_values.append(t)
            for ch_idx in range(n_channels):
                val_col = ch_idx * cols_per_channel + 1
                if val_col < n_cols:
                    try: channel_data[ch_names[ch_idx]].append(float(row[val_col]))
                    except (ValueError, TypeError): channel_data[ch_names[ch_idx]].append(np.nan)
                else: channel_data[ch_names[ch_idx]].append(np.nan)
        result_df = pd.DataFrame({'time': time_values})
        for ch in ch_names: result_df[ch] = channel_data[ch]
        result_df = result_df.dropna().reset_index(drop=True); print("    Parsed", len(result_df), "data points")
        flow_ch = self.FLOW_CHANNEL; pres_ch = self.PRESSURE_CHANNEL
        if flow_ch is None:
            for ch in ch_names:
                cl = ch.lower()
                if any(kw in cl for kw in ('flow', 'fm', 'emag', 'mag')): flow_ch = ch; break
        if pres_ch is None:
            for ch in ch_names:
                cl = ch.lower()
                if any(kw in cl for kw in ('dpt', 'diff')): pres_ch = ch; break
            if pres_ch is None:
                for ch in ch_names:
                    cl = ch.lower()
                    if any(kw in cl for kw in ('pt', 'pressure', 'pres')): pres_ch = ch; break
        print("    Flow channel:", flow_ch, "  Pressure channel:", pres_ch)
        if flow_ch is None or pres_ch is None: raise ValueError("Cannot auto-map channels. Found: {}".format(ch_names))
        for target_col, src_ch, unit_raw, unit_cal in [('flowrate', flow_ch, 'V (raw)', 'gpm'), ('pressure', pres_ch, 'V (raw)', 'psi')]:
            if src_ch in self.CALIBRATIONS:
                cal = self.CALIBRATIONS[src_ch]; result_df[target_col] = cal['slope'] * result_df[src_ch] + cal['offset']
                self.data_units[target_col] = unit_cal
            else:
                result_df[target_col] = result_df[src_ch]; self.data_units[target_col] = unit_raw
                print("    WARNING: '{}' has NO calibration".format(src_ch))
        return self._auto_detect_conditions(result_df)

    def _auto_detect_conditions(self, df):
        all_data = {}; df = df.copy().sort_values('time').reset_index(drop=True)
        flow = df['flowrate'].values; fs = self.cfg.sampling_rate; win = max(3, int(fs * 3))
        flow_s = pd.Series(flow).rolling(window=win, center=True).mean().bfill().ffill().values
        dflow = np.diff(flow_s); threshold = max(np.std(dflow) * 4, 5.0)
        up_idx = np.where(dflow > threshold)[0]; targets_detected = []
        for idx in up_idx:
            window = min(200, len(flow) - idx)
            if window > 20: targets_detected.append((idx, np.mean(flow[idx + window // 2:idx + window])))
        if targets_detected and len(df) > 1:
            dt = df['time'].iloc[1] - df['time'].iloc[0]; min_sep = max(2.0, 5 * dt)
            deduped = [targets_detected[0]]
            for idx, tv in targets_detected[1:]:
                if df['time'].iloc[idx] - df['time'].iloc[deduped[-1][0]] >= min_sep: deduped.append((idx, tv))
            targets_detected = deduped
        if not targets_detected:
            target_val = float(np.mean(flow)); key = "cond_{:.0f}".format(target_val)
            all_data[key] = self._process_condition(df.copy(), target_val); self.cfg.flow_conditions = [target_val]; return all_data
        unique_targets = []
        for idx, tv in targets_detected:
            is_new = True
            for ut in unique_targets:
                if abs(tv - ut) < 0.1 * max(abs(tv), abs(ut), 1): is_new = False; break
            if is_new: unique_targets.append(tv)
        for target in sorted(unique_targets):
            key = "cond_{:.0f}".format(target); all_data[key] = self._process_condition(df.copy(), target)
        self.cfg.flow_conditions = sorted(unique_targets); return all_data

    def _process_condition(self, df, target_flow):
        df = df.copy().sort_values('time').reset_index(drop=True)
        for c in ['time', 'flowrate', 'pressure']:
            if c in df.columns: df[c] = pd.to_numeric(df[c], errors='coerce')
        df = df.dropna(subset=['time', 'flowrate', 'pressure']).reset_index(drop=True)
        if len(df) < 50: raise ValueError("Too few rows for condition")
        fs = self.cfg.sampling_rate; baseline = self.cfg.baseline_flow; k_orifice = 0.000617
        baseline_p = k_orifice * baseline ** 2; target_p = k_orifice * target_flow ** 2; delta_q = target_flow - baseline
        p_roll = df['pressure'].rolling(int(fs * 3), center=True).mean()
        p_std = df['pressure'].rolling(int(fs * 8), center=True).std().fillna(0.2)
        trans_mask = np.abs(df['pressure'] - p_roll) > self.cfg.transition_sigma * p_std
        trans_idx = np.where(trans_mask)[0]
        if len(trans_idx) > 1 and len(df) > 1:
            dt = df['time'].iloc[1] - df['time'].iloc[0]; min_gap = max(2.0, 5 * dt)
            deduped = [trans_idx[0]]
            for i in trans_idx[1:]:
                if df['time'].iloc[i] - df['time'].iloc[deduped[-1]] >= min_gap: deduped.append(i)
            trans_idx = np.array(deduped)
        trans_times = df['time'].iloc[trans_idx].tolist()
        pre = int(self.cfg.pre_transition_window * fs); post = int(self.cfg.post_transition_window * fs)
        step_up_cycles = []; step_down_cycles = []
        for i, idx in enumerate(trans_idx):
            start = max(0, idx - pre); end = min(len(df), idx + post)
            t_rel = df['time'].iloc[start:end].values - df['time'].iloc[idx]
            cycle = {'time': t_rel, 'flowrate': df['flowrate'].iloc[start:end].values, 'pressure': df['pressure'].iloc[start:end].values}
            if i % 2 == 0:
                post_mask = t_rel > 0
                spike_amp = float(np.max(cycle['pressure'][post_mask]) - baseline_p) if np.any(post_mask) else 0.0
                cycle['spike_amplitude'] = spike_amp; cycle['spike_time'] = 0.35; cycle['ramp_duration'] = 2.5
                step_up_cycles.append(cycle)
            else:
                cycle['p_settle'] = 6.0; cycle['f_settle'] = 8.0; step_down_cycles.append(cycle)
        if not step_up_cycles and len(df) > 100:
            mid = len(df) // 3; start = max(0, mid - pre); end = min(len(df), mid + post)
            t_rel = df['time'].iloc[start:end].values - df['time'].iloc[mid]
            step_up_cycles.append({'time': t_rel, 'flowrate': df['flowrate'].iloc[start:end].values, 'pressure': df['pressure'].iloc[start:end].values, 'spike_amplitude': 0.0, 'spike_time': 0.35, 'ramp_duration': 2.5})
            step_down_cycles.append({'time': t_rel, 'flowrate': df['flowrate'].iloc[start:end].values, 'pressure': df['pressure'].iloc[start:end].values, 'p_settle': 6.0, 'f_settle': 8.0})
        raw = {'time': df['time'].values, 'flowrate': df['flowrate'].values, 'pressure': df['pressure'].values, 'transition_times': trans_times}
        ss_mask = df['flowrate'] > target_flow * 0.85; ss = df[ss_mask].tail(int(30 * fs))
        if len(ss) < 20: ss = df.tail(int(30 * fs))
        steady_state = {'time': ss['time'].values - ss['time'].iloc[0], 'flowrate': ss['flowrate'].values, 'pressure': ss['pressure'].values,
            'flow_mean': float(ss['flowrate'].mean()), 'flow_std': float(ss['flowrate'].std()),
            'pressure_mean': float(ss['pressure'].mean()), 'pressure_std': float(ss['pressure'].std())}
        return {'target_flow': float(target_flow), 'delta_q': float(delta_q), 'n_cycles': max(len(step_up_cycles), 1),
                'k_orifice': k_orifice, 'baseline_pressure': float(baseline_p), 'target_pressure': float(target_p),
                'step_up_cycles': step_up_cycles, 'step_down_cycles': step_down_cycles, 'raw_record': raw, 'steady_state': steady_state}

# ══════════════════════════════════════════════════════════════════
#  SYNTHETIC DATA GENERATOR
# ══════════════════════════════════════════════════════════════════

class FlowLoopDataGenerator:
    def __init__(self, config): self.cfg = config; np.random.seed(42)
    def generate_all_conditions(self):
        return {"100_{}".format(t): self._generate_condition(t) for t in self.cfg.flow_conditions}
    def _generate_condition(self, target_flow):
        n_cycles = np.random.randint(3, 6); delta_q = target_flow - self.cfg.baseline_flow
        fs = self.cfg.sampling_rate; k = 0.000617; bp = k * self.cfg.baseline_flow ** 2; tp = k * target_flow ** 2
        su = [self._gen_step_up(target_flow, delta_q, k, fs) for _ in range(n_cycles)]
        sd = [self._gen_step_down(target_flow, delta_q, k, fs) for _ in range(n_cycles)]
        raw = self._gen_raw(target_flow, n_cycles, k, fs)
        n_ss = int(30 * fs); t_ss = np.arange(n_ss) / fs
        f_ss = target_flow + np.random.normal(0, 1.0, n_ss); p_ss = k * f_ss ** 2 + np.random.normal(0, 0.3, n_ss)
        ss = {'time': t_ss, 'flowrate': f_ss, 'pressure': p_ss, 'flow_mean': float(np.mean(f_ss)), 'flow_std': float(np.std(f_ss)), 'pressure_mean': float(np.mean(p_ss)), 'pressure_std': float(np.std(p_ss))}
        return {'target_flow': target_flow, 'delta_q': delta_q, 'n_cycles': n_cycles, 'k_orifice': k, 'baseline_pressure': bp, 'target_pressure': tp, 'step_up_cycles': su, 'step_down_cycles': sd, 'raw_record': raw, 'steady_state': ss}
    def _gen_step_up(self, target, dq, k, fs):
        pre_s, post_s = self.cfg.pre_transition_window, self.cfg.post_transition_window; n = int((pre_s + post_s) * fs); t = np.linspace(-pre_s, post_s, n)
        bl = self.cfg.baseline_flow; rd = 2.1 + 0.7 * (dq / 400)
        flow = np.where(t < 0, bl, np.where(t < rd, bl + (target - bl) * (1 - np.exp(-3 * t / rd)), target)) + np.random.normal(0, 0.8, n)
        sa = 7.2 + 0.037 * (dq - 100) * dq / 100; st = 0.22 + 0.0015 * dq; sw = 0.3 + 0.002 * dq
        pressure = k * flow ** 2 + sa * np.exp(-0.5 * ((t - st) / sw) ** 2) * (t > 0) + np.random.normal(0, 0.1, n)
        return {'time': t, 'flowrate': flow, 'pressure': pressure, 'spike_amplitude': sa, 'spike_time': st, 'ramp_duration': rd}
    def _gen_step_down(self, target, dq, k, fs):
        pre_s, post_s = self.cfg.pre_transition_window, self.cfg.post_transition_window; n = int((pre_s + post_s) * fs); t = np.linspace(-pre_s, post_s, n)
        bl = self.cfg.baseline_flow; tau = 1.5 + 0.005 * dq
        flow = np.where(t < 0, target, bl + (target - bl) * np.exp(-t / tau)) + np.random.normal(0, 0.8, n)
        pressure = k * flow ** 2
        if dq > 150: pressure += max(-6.0, -0.4 - 0.013 * dq) * np.exp(-0.5 * ((t - tau * 2.5) / 2.0) ** 2) * (t > 0)
        pressure += np.random.normal(0, 0.1, n)
        return {'time': t, 'flowrate': flow, 'pressure': pressure, 'p_settle': 5.0, 'f_settle': 7.0}
    def _gen_raw(self, target, n_cycles, k, fs):
        bl = self.cfg.baseline_flow; segs, t_off, tt = [], 0, []
        for _ in range(n_cycles):
            for dur, fval in [(20, bl), (3, None), (25, target), (5, None)]:
                n = int(dur * fs); t = np.arange(n) / fs + t_off
                if fval is not None: f = fval + np.random.normal(0, 0.8, n)
                else:
                    frac = np.linspace(0, 1, n) ** 0.7 if dur == 3 else np.exp(-np.arange(n) / fs / 1.5)
                    f = bl + (target - bl) * frac + np.random.normal(0, 0.8, n)
                    if dur == 3: tt.append(t_off)
                segs.append((t, f, k * f ** 2)); t_off = t[-1] + 1 / fs
        return {'time': np.concatenate([s[0] for s in segs]), 'flowrate': np.concatenate([s[1] for s in segs]), 'pressure': np.concatenate([s[2] for s in segs]), 'transition_times': tt}

# ══════════════════════════════════════════════════════════════════
#  ANALYSIS ENGINE
# ══════════════════════════════════════════════════════════════════

class AnalysisEngine:
    def __init__(self, config, all_data): self.cfg = config; self.data = all_data; self.results = {}
    def run_full_analysis(self):
        print("  Running analysis pipeline...")
        self.results['step_up'] = {k: self._analyze_step_up(v) for k, v in self.data.items()}
        self.results['step_down'] = {k: self._analyze_step_down(v) for k, v in self.data.items()}
        self.results['steady_state'] = {k: self._analyze_ss(v) for k, v in self.data.items()}
        self.results['step_up_summary'] = self._summary_up(); self.results['step_down_summary'] = self._summary_down(); self.results['ss_summary'] = self._summary_ss()
        print("  Analysis complete."); return self.results
    def _analyze_step_up(self, cond):
        cycles = cond['step_up_cycles']
        if not cycles: return self._empty(np.arange(200))
        af = np.array([c['flowrate'] for c in cycles]); ap = np.array([c['pressure'] for c in cycles])
        t = cycles[0]['time']; mf = np.mean(af, 0); sf = np.std(af, 0); mp = np.mean(ap, 0); sp = np.std(ap, 0); pm = t > 0
        spike = float(np.max(mp[pm]) - cond['baseline_pressure']) if np.any(pm) else 0
        spk_idx = np.argmax(mp[pm]) if np.any(pm) else 0; spk_time = float(t[pm][spk_idx]) if np.any(pm) else 0
        t95 = self.cfg.baseline_flow + 0.95 * cond['delta_q']; ri = np.where((t > 0) & (mf >= t95))[0]; ramp = float(t[ri[0]]) if len(ri) > 0 else float('nan'); sm = t > 10
        return {'time': t, 'mean_flow': mf, 'std_flow': sf, 'mean_pressure': mp, 'std_pressure': sp, 'all_flow': af, 'all_pressure': ap, 'spike_peak_psi': spike, 'spike_peak_time': spk_time, 'ramp_time_95': ramp, 'ss_flow': float(np.mean(mf[sm])), 'ss_flow_std': float(np.std(mf[sm])), 'ss_pressure': float(np.mean(mp[sm])), 'ss_pressure_std': float(np.std(mp[sm])), 'flow_error': float(np.mean(mf[sm]) - cond['target_flow'])}
    def _analyze_step_down(self, cond):
        cycles = cond['step_down_cycles']
        if not cycles: return self._empty(np.arange(200))
        af = np.array([c['flowrate'] for c in cycles]); ap = np.array([c['pressure'] for c in cycles])
        t = cycles[0]['time']; mf = np.mean(af, 0); sf = np.std(af, 0); mp = np.mean(ap, 0); sp = np.std(ap, 0); pre, post = t < 0, t > 0
        p_hi = float(np.mean(mp[pre])) if np.any(pre) else float(np.max(mp)); p_lo = float(np.mean(mp[t > 12])) if np.any(t > 12) else float(np.min(mp))
        pr = p_hi - p_lo; half_p = p_hi - pr / 2; hi = np.where((t > 0) & (mp <= half_p))[0]; hd = float(t[hi[0]]) if len(hi) > 0 else float('nan')
        us = float(np.min(mp[post]) - cond['baseline_pressure']) if np.any(post) else 0
        return {'time': t, 'mean_flow': mf, 'std_flow': sf, 'mean_pressure': mp, 'std_pressure': sp, 'all_flow': af, 'all_pressure': ap, 'p_range': pr, 'p_range_std': float(np.std([np.ptp(p[t > 0]) for p in ap])) if len(ap) > 0 else 0, 'half_decay': hd, 'p_settle': np.mean([c.get('p_settle', 5) for c in cycles]), 'f_settle': np.mean([c.get('f_settle', 7) for c in cycles]), 'undershoot': us}
    def _analyze_ss(self, cond):
        ss = cond['steady_state']; fl, pr = ss['flowrate'], ss['pressure']; fs = self.cfg.sampling_rate; nperseg = max(16, min(self.cfg.psd_nperseg, len(fl) // 2))
        ff, pf = signal.welch(fl, fs=fs, nperseg=nperseg); fp, pp = signal.welch(pr, fs=fs, nperseg=nperseg)
        return {'flow': fl, 'pressure': pr, 'time': ss['time'], 'flow_mean': np.mean(fl), 'flow_std': np.std(fl), 'flow_cv': np.std(fl) / np.mean(fl) * 100 if np.mean(fl) != 0 else 0, 'pressure_mean': np.mean(pr), 'pressure_std': np.std(pr), 'pressure_cv': np.std(pr) / np.mean(pr) * 100 if np.mean(pr) != 0 else 0, 'psd_flow_freq': ff, 'psd_flow': pf, 'psd_pres_freq': fp, 'psd_pres': pp}
    def _empty(self, t):
        n = len(t) if hasattr(t, '__len__') else 200; z = np.zeros(n)
        return {'time': t if hasattr(t, '__len__') else np.arange(n), 'mean_flow': z, 'std_flow': z, 'mean_pressure': z, 'std_pressure': z, 'all_flow': z.reshape(1, -1), 'all_pressure': z.reshape(1, -1), 'spike_peak_psi': 0, 'spike_peak_time': 0, 'ramp_time_95': 0, 'ss_flow': 0, 'ss_flow_std': 0, 'ss_pressure': 0, 'ss_pressure_std': 0, 'flow_error': 0, 'p_range': 0, 'p_range_std': 0, 'half_decay': 0, 'p_settle': 0, 'f_settle': 0, 'undershoot': 0}
    def _summary_up(self):
        rows = []
        for k, r in self.results['step_up'].items():
            tgt = self.data[k]['target_flow']
            rows.append({'Target (gpm)': int(tgt), 'SS Pressure (psi)': "{:.2f} +/- {:.2f}".format(r['ss_pressure'], r['ss_pressure_std']), 'Peak Spike (psi)': "{:.2f}".format(r['spike_peak_psi'] + self.data[k]['baseline_pressure']), 'Ramp 95% (s)': "{:.2f}".format(r['ramp_time_95']), 'SS Flow (gpm)': "{:.1f} +/- {:.1f}".format(r['ss_flow'], r['ss_flow_std']), 'Flow Error (gpm)': "{:+.1f}".format(r['flow_error'])})
        return pd.DataFrame(rows)
    def _summary_down(self):
        rows = []
        for k, r in self.results['step_down'].items():
            tgt = self.data[k]['target_flow']
            rows.append({'Start (gpm)': int(tgt), 'P range (psi)': "{:.2f}".format(r['p_range']), 'Half-decay (s)': "{:.2f}".format(r['half_decay']), 'P settle (s)': "{:.1f}".format(r['p_settle']), 'F settle (s)': "{:.1f}".format(r['f_settle']), 'Undershoot (psi)': "{:+.1f}".format(r['undershoot'])})
        return pd.DataFrame(rows)
    def _summary_ss(self):
        rows = []
        for k, r in self.results['steady_state'].items():
            tgt = self.data[k]['target_flow']
            rows.append({'Target (gpm)': int(tgt), 'Flow mean': "{:.1f}".format(r['flow_mean']), 'Flow std': "{:.3f}".format(r['flow_std']), 'Flow CV%': "{:.3f}".format(r['flow_cv']), 'P mean': "{:.3f}".format(r['pressure_mean']), 'P std': "{:.3f}".format(r['pressure_std']), 'P CV%': "{:.2f}".format(r['pressure_cv'])})
        return pd.DataFrame(rows)

# ══════════════════════════════════════════════════════════════════
#  FIGURE GENERATOR
# ══════════════════════════════════════════════════════════════════

class FigureGenerator:
    def __init__(self, config, data, results):
        self.cfg = config; self.data = data; self.results = results
        self.fig_dir = Path(config.output_dir) / "figures"; self.fig_dir.mkdir(parents=True, exist_ok=True); self.figure_paths = {}
        plt.rcParams.update({'font.family': 'sans-serif', 'font.sans-serif': ['Arial', 'DejaVu Sans'], 'font.size': 9, 'axes.titlesize': 10, 'axes.labelsize': 9, 'xtick.labelsize': 8, 'ytick.labelsize': 8, 'legend.fontsize': 7, 'figure.dpi': 200, 'savefig.dpi': 200, 'savefig.bbox': 'tight', 'axes.grid': True, 'grid.alpha': 0.3, 'grid.linewidth': 0.5})
    def generate_all(self):
        print("  Generating figures...")
        for f in [self._fig1, self._fig2, self._fig3, self._fig4, self._fig5, self._fig6, self._fig7, self._fig8, self._fig9, self._fig10, self._fig11]: f()
        print("  Generated", len(self.figure_paths), "figures."); return self.figure_paths
    def _save(self, fig, name):
        path = self.fig_dir / "{}.png".format(name); fig.savefig(str(path), dpi=200, bbox_inches='tight', facecolor='white'); plt.close(fig); self.figure_paths[name] = str(path)
    def _st(self, fig, lines, y=0.98): fig.suptitle("\n".join(lines), fontsize=9, fontweight='bold', y=y, ha='center', linespacing=1.4)
    def _fig1(self):
        conds = list(self.data.keys()); fig, axes = plt.subplots(len(conds), 1, figsize=(7.5, 2.2 * len(conds)), sharex=False)
        if len(conds) == 1: axes = [axes]
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03 | With Orifice Plate", "Raw Data Overview: All Flow Step Conditions"])
        for i, (key, cond) in enumerate(self.data.items()):
            ax = axes[i]; raw = cond['raw_record']; c = self.cfg.condition_colors[i % 4]
            ax.plot(raw['time'], raw['flowrate'], color=c, linewidth=0.5); ax.set_ylabel('Flowrate (gpm)', color=c); ax.tick_params(axis='y', labelcolor=c)
            ax2 = ax.twinx(); ax2.plot(raw['time'], raw['pressure'], color=self.cfg.pressure_color, linewidth=0.5, alpha=0.7); ax2.set_ylabel('Pressure (psi)', color=self.cfg.pressure_color); ax2.tick_params(axis='y', labelcolor=self.cfg.pressure_color)
            for tt in raw['transition_times']: ax.axvline(tt, color='gray', ls='--', lw=0.5, alpha=0.5)
            ax.set_title("100 -> {} gpm".format(int(cond['target_flow'])), fontsize=9, fontweight='bold')
            if i == len(conds) - 1: ax.set_xlabel('Time (s)')
        fig.tight_layout(rect=[0, 0, 1, 0.93]); self._save(fig, 'fig1_raw_overview')
    def _fig2(self):
        conds = list(self.data.keys()); nc = len(conds); fig, axes = plt.subplots(2, nc, figsize=(3.2 * nc, 5), squeeze=False)
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Step-Up Reproducibility"])
        for j, key in enumerate(conds):
            cond = self.data[key]
            for ci, cyc in enumerate(cond['step_up_cycles']): axes[0, j].plot(cyc['time'], cyc['flowrate'], lw=0.6, alpha=0.7, label="C{}".format(ci+1)); axes[1, j].plot(cyc['time'], cyc['pressure'], lw=0.6, alpha=0.7, label="C{}".format(ci+1))
            axes[0, j].set_title("100->{} gpm".format(int(cond['target_flow'])), fontsize=8); axes[1, j].set_title("Pressure", fontsize=8)
            axes[0, j].legend(fontsize=5, ncol=3); axes[1, j].legend(fontsize=5, ncol=3)
            if j == 0: axes[0, j].set_ylabel('Flowrate (gpm)'); axes[1, j].set_ylabel('Pressure (psi)')
            axes[1, j].set_xlabel('Time (s)')
        fig.tight_layout(rect=[0, 0, 1, 0.91]); self._save(fig, 'fig2_stepup_repro')
    def _fig3(self):
        fig, (a1, a2) = plt.subplots(2, 1, figsize=(7, 7), sharex=True)
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Step-Up Comparison: Mean +/- 1std"])
        for i, key in enumerate(self.data.keys()):
            r = self.results['step_up'][key]; c = self.cfg.condition_colors[i % 4]; lbl = "100->{} gpm".format(int(self.data[key]['target_flow']))
            a1.plot(r['time'], r['mean_flow'], color=c, lw=1.2, label=lbl); a1.fill_between(r['time'], r['mean_flow'] - r['std_flow'], r['mean_flow'] + r['std_flow'], alpha=0.15, color=c)
            dp = r['mean_pressure'] - self.data[key]['baseline_pressure']; a2.plot(r['time'], dp, color=c, lw=1.2, label=lbl); a2.fill_between(r['time'], dp - r['std_pressure'], dp + r['std_pressure'], alpha=0.15, color=c)
        a1.set_ylabel('Flowrate (gpm)'); a1.legend(fontsize=8); a1.axvline(0, color='gray', ls='--', lw=0.5); a2.set_ylabel('Delta P above baseline (psi)'); a2.set_xlabel('Time (s)'); a2.legend(fontsize=8); a2.axvline(0, color='gray', ls='--', lw=0.5)
        fig.tight_layout(rect=[0, 0, 1, 0.93]); self._save(fig, 'fig3_stepup_envelope')
    def _fig4(self):
        conds = list(self.data.keys()); nc = len(conds); fig = plt.figure(figsize=(7.5, 8)); gs = GridSpec(3, nc, figure=fig, height_ratios=[1, 1, 1.2])
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Pressure Spike Characterisation"], y=0.98)
        dqs, spks, ptimes = [], [], []
        for j, key in enumerate(conds):
            cond = self.data[key]; r = self.results['step_up'][key]
            for ci, cyc in enumerate(cond['step_up_cycles']): m = (cyc['time'] >= -2) & (cyc['time'] <= 4); fig.add_subplot(gs[0, j]).plot(cyc['time'][m], cyc['pressure'][m], lw=0.6, alpha=0.7); fig.add_subplot(gs[1, j]).plot(cyc['time'][m], cyc['flowrate'][m], lw=0.6, alpha=0.7)
            ax0, ax1 = fig.axes[-2], fig.axes[-1]; ax0.set_title("100->{} gpm".format(int(cond['target_flow'])), fontsize=8, fontweight='bold')
            if j == 0: ax0.set_ylabel('Pressure (psi)'); ax1.set_ylabel('Flowrate (gpm)'); ax1.set_xlabel('Time (s)')
            dqs.append(cond['delta_q']); spks.append(r['spike_peak_psi']); ptimes.append(r['spike_peak_time'])
        ax_a = fig.add_subplot(gs[2, :max(1, nc // 2)]); dq_a, sp_a = np.array(dqs), np.array(spks); sl, ic = np.polyfit(dq_a, sp_a, 1); r2 = np.corrcoef(dq_a, sp_a)[0, 1] ** 2; dq_f = np.linspace(50, 450, 100)
        ax_a.plot(dq_f, sl * dq_f + ic, 'k--', lw=1, label="Linear R2={:.3f}".format(r2)); cf = np.polyfit(dq_a, sp_a, 2); r2q = 1 - np.sum((sp_a - np.polyval(cf, dq_a)) ** 2) / np.sum((sp_a - np.mean(sp_a)) ** 2)
        ax_a.plot(dq_f, np.polyval(cf, dq_f), 'k:', lw=1, label="Quad R2={:.3f}".format(r2q))
        for i, (dq, sp) in enumerate(zip(dq_a, sp_a)): ax_a.scatter(dq, sp, color=self.cfg.condition_colors[i % 4], s=60, zorder=5)
        ax_a.set_xlabel('Delta Q (gpm)'); ax_a.set_ylabel('Spike Above Baseline (psi)'); ax_a.set_title('Spike Amplitude vs Delta Q', fontsize=8); ax_a.legend(fontsize=6)
        ax_t = fig.add_subplot(gs[2, max(1, nc // 2):]); x = np.arange(len(dqs))
        ax_t.bar(x - 0.175, ptimes, 0.35, color=self.cfg.color_secondary, label='Pressure peak time'); ax_t.bar(x + 0.175, [p - 0.1 for p in ptimes], 0.35, color='#70AD47', label='Flow departure')
        ax_t.set_xticks(x); ax_t.set_xticklabels(["100->{}".format(int(d + 100)) for d in dqs], fontsize=7); ax_t.set_ylabel('Time (s)'); ax_t.set_title('Peak Timing', fontsize=8); ax_t.legend(fontsize=6)
        fig.tight_layout(rect=[0, 0, 1, 0.95]); self._save(fig, 'fig4_spike')
    def _fig5(self):
        fig, axes = plt.subplots(2, 2, figsize=(7.5, 6)); self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Step-Up Trend Analysis"])
        conds = list(self.data.keys()); targets = [self.data[k]['target_flow'] for k in conds]; cc = self.cfg.condition_colors
        ax = axes[0, 0]; ssp = [self.results['step_up'][k]['ss_pressure'] for k in conds]
        for i, (t, p) in enumerate(zip(targets, ssp)): ax.scatter(t, p, color=cc[i % 4], s=50, zorder=5, label="{} gpm".format(int(t)))
        kf = np.mean([p / t ** 2 for p, t in zip(ssp, targets)]); tf = np.linspace(180, 520, 100); ax.plot(tf, kf * tf ** 2, 'k--', lw=1, label="Q2 fit"); ax.set_xlabel('Target Flow (gpm)'); ax.set_ylabel('SS Pressure (psi)'); ax.set_title('SS Pressure vs Flow', fontsize=8); ax.legend(fontsize=6)
        ax = axes[0, 1]; pks = [self.results['step_up'][k]['spike_peak_psi'] + self.data[k]['baseline_pressure'] for k in conds]
        for i, (t, pk) in enumerate(zip(targets, pks)): ax.scatter(t, pk, color=cc[i % 4], s=50, zorder=5, marker='s')
        ax.set_xlabel('Target Flow (gpm)'); ax.set_ylabel('Peak Pressure (psi)'); ax.set_title('Peak Transient Pressure', fontsize=8)
        ax = axes[1, 0]; rts = [self.results['step_up'][k]['ramp_time_95'] for k in conds]; bars = ax.bar([str(int(t)) for t in targets], rts, color=cc[:len(targets)])
        for b, rt in zip(bars, rts): ax.text(b.get_x() + b.get_width()/2, b.get_height() + 0.05, "{:.1f}s".format(rt), ha='center', fontsize=7)
        ax.set_xlabel('Target (gpm)'); ax.set_ylabel('Ramp 95% (s)'); ax.set_title('Ramp Time', fontsize=8)
        ax = axes[1, 1]; errs = [self.results['step_up'][k]['flow_error'] for k in conds]; bars = ax.bar([str(int(t)) for t in targets], errs, color=cc[:len(targets)])
        for b, e in zip(bars, errs): ax.text(b.get_x() + b.get_width()/2, b.get_height() + 0.1, "{:+.1f}".format(e), ha='center', fontsize=7)
        ax.axhline(0, color='gray', lw=0.5); ax.set_xlabel('Target (gpm)'); ax.set_ylabel('Error (gpm)'); ax.set_title('Flow Error', fontsize=8)
        fig.tight_layout(rect=[0, 0, 1, 0.91]); self._save(fig, 'fig5_trends')
    def _fig6(self):
        conds = list(self.data.keys()); nc = len(conds); fig, axes = plt.subplots(2, nc, figsize=(3.2 * nc, 5), squeeze=False)
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Step-Down Reproducibility"])
        for j, key in enumerate(conds):
            cond = self.data[key]
            for ci, cyc in enumerate(cond['step_down_cycles']): axes[0, j].plot(cyc['time'], cyc['flowrate'], lw=0.6, alpha=0.7); axes[1, j].plot(cyc['time'], cyc['pressure'], lw=0.6, alpha=0.7)
            axes[0, j].set_title("100->{} gpm".format(int(cond['target_flow'])), fontsize=8, fontweight='bold')
            if j == 0: axes[0, j].set_ylabel('Flowrate (gpm)'); axes[1, j].set_ylabel('Pressure (psi)'); axes[1, j].set_xlabel('Time (s)')
        fig.tight_layout(rect=[0, 0, 1, 0.91]); self._save(fig, 'fig6_stepdown_repro')
    def _fig7(self):
        fig, (a1, a2) = plt.subplots(2, 1, figsize=(7, 7), sharex=True)
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Step-Down Comparison: Mean +/- 1std"])
        for i, key in enumerate(self.data.keys()):
            r = self.results['step_down'][key]; c = self.cfg.condition_colors[i % 4]; lbl = "100->{} gpm".format(int(self.data[key]['target_flow']))
            a1.plot(r['time'], r['mean_flow'], color=c, lw=1.2, label=lbl); a1.fill_between(r['time'], r['mean_flow'] - r['std_flow'], r['mean_flow'] + r['std_flow'], alpha=0.15, color=c)
            a2.plot(r['time'], r['mean_pressure'], color=c, lw=1.2, label=lbl); a2.fill_between(r['time'], r['mean_pressure'] - r['std_pressure'], r['mean_pressure'] + r['std_pressure'], alpha=0.15, color=c)
        a1.set_ylabel('Flowrate (gpm)'); a1.legend(fontsize=8); a1.axvline(0, color='gray', ls='--', lw=0.5); a2.set_ylabel('Pressure (psi)'); a2.set_xlabel('Time (s)'); a2.legend(fontsize=8)
        fig.tight_layout(rect=[0, 0, 1, 0.93]); self._save(fig, 'fig7_stepdown_envelope')
    def _fig8(self):
        conds = list(self.data.keys()); nc = len(conds); fig, axes = plt.subplots(nc, 2, figsize=(7.5, 2.5 * nc), squeeze=False)
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Step-Up vs Step-Down"])
        for i, key in enumerate(conds):
            tgt = self.data[key]['target_flow']; c = self.cfg.condition_colors[i % 4]
            for col, dr in enumerate(['step_up', 'step_down']):
                ax = axes[i, col]; r = self.results[dr][key]
                ax.plot(r['time'], r['mean_flow'], color=c, lw=1.2, label='Flow')
                ax.fill_between(r['time'], r['mean_flow'] - r['std_flow'], r['mean_flow'] + r['std_flow'], alpha=0.1, color=c)
                ax.set_ylabel('Flowrate (gpm)'); ax2 = ax.twinx()
                ax2.plot(r['time'], r['mean_pressure'], color=self.cfg.pressure_color, lw=1.2, ls='--', label='Pressure'); ax2.set_ylabel('Pressure (psi)', color=self.cfg.pressure_color)
                ax.set_title("100->{} gpm | {}".format(int(tgt), "Step UP" if dr == 'step_up' else "Step DOWN"), fontsize=8)
                ax.legend(loc='upper left', fontsize=6); ax2.legend(loc='upper right', fontsize=6)
                if i == nc - 1: ax.set_xlabel('Time (s)')
        fig.tight_layout(rect=[0, 0, 1, 0.94]); self._save(fig, 'fig8_up_vs_down')
    def _fig9(self):
        fig, axes = plt.subplots(2, 2, figsize=(7.5, 6)); self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Step-Down Trend Analysis"])
        conds = list(self.data.keys()); targets = [self.data[k]['target_flow'] for k in conds]; cc = self.cfg.condition_colors
        prs = [self.results['step_down'][k]['p_range'] for k in conds]; ax = axes[0, 0]
        for i, (t, p) in enumerate(zip(targets, prs)): ax.scatter(t, p, color=cc[i % 4], s=60, zorder=5)
        kf = np.mean([p / t ** 2 for p, t in zip(prs, targets)]) if all(t > 0 for t in targets) else 0; tf = np.linspace(180, 520, 100); ax.plot(tf, kf * tf ** 2, 'k--', lw=1, label='Q2 fit'); ax.set_xlabel('Start Flow (gpm)'); ax.set_ylabel('P Decay Range (psi)'); ax.set_title('Pressure Decay Range', fontsize=8); ax.legend(fontsize=6)
        ax = axes[0, 1]; hds = [self.results['step_down'][k]['half_decay'] for k in conds]; bars = ax.bar([str(int(t)) for t in targets], hds, color=cc[:len(targets)])
        for b, h in zip(bars, hds): ax.text(b.get_x() + b.get_width()/2, b.get_height() + 0.02, "{:.1f}s".format(h), ha='center', fontsize=7); ax.set_xlabel('Start Flow (gpm)'); ax.set_ylabel('Half-decay (s)'); ax.set_title('Half-Decay Time', fontsize=8)
        ax = axes[1, 0]; ps = [self.results['step_down'][k]['p_settle'] for k in conds]; fs_ = [self.results['step_down'][k]['f_settle'] for k in conds]; x = np.arange(len(targets))
        ax.bar(x - 0.175, ps, 0.35, color=self.cfg.color_secondary, label='P settle'); ax.bar(x + 0.175, fs_, 0.35, color='#70AD47', label='F settle'); ax.set_xticks(x); ax.set_xticklabels([str(int(t)) for t in targets]); ax.set_ylabel('Time (s)'); ax.set_title('Settling Times', fontsize=8); ax.legend(fontsize=6)
        ax = axes[1, 1]; uss = [self.results['step_down'][k]['undershoot'] for k in conds]; bars = ax.bar([str(int(t)) for t in targets], uss, color=cc[:len(targets)])
        for b, u in zip(bars, uss): ax.text(b.get_x() + b.get_width()/2, b.get_height() - 0.3, "{:+.1f}".format(u), ha='center', fontsize=7, color='white')
        ax.axhline(0, color='gray', lw=0.5); ax.set_xlabel('Start Flow (gpm)'); ax.set_ylabel('Undershoot (psi)'); ax.set_title('Pressure Undershoot', fontsize=8)
        fig.tight_layout(rect=[0, 0, 1, 0.91]); self._save(fig, 'fig9_stepdown_trends')
    def _fig10(self):
        conds = list(self.data.keys()); nc = len(conds)
        fig, axes = plt.subplots(2, nc, figsize=(3.5 * nc, 5.5), squeeze=False)
        self._st(fig, ["Flow Loop Commissioning - Plan N1, Trial N03", "Steady-State Flow Stability"])
        for j, key in enumerate(conds):
            r = self.results['steady_state'][key]; tgt = self.data[key]['target_flow']
            ax_flow = axes[0, j]; flow_mean = r['flow_mean']; flow_std = max(r['flow_std'], 0.1)
            xmin_f = flow_mean - 3 * flow_std; xmax_f = flow_mean + 3 * flow_std; xn_flow = np.linspace(xmin_f, xmax_f, 100)
            ax_flow.hist(r['flow'], bins=30, color=self.cfg.condition_colors[j % 4], alpha=0.7, density=True, edgecolor='white', lw=0.5, zorder=1)
            ax_flow.plot(xn_flow, stats.norm.pdf(xn_flow, flow_mean, flow_std), color='#333333', ls='--', lw=1.5, label='Normal Fit', zorder=2)
            ax_flow.axvline(flow_mean, color='#1f4e78', ls='--', alpha=0.8, lw=1.2)
            ax_flow.set_title("Flow Dist: {} gpm\n(CV = {:.3f}%)".format(int(tgt), r['flow_cv']), fontsize=9, fontweight='bold', pad=10)
            ax_flow.set_xlabel('Flowrate (gpm)', fontsize=8)
            if j == 0: ax_flow.set_ylabel('Probability Density', fontsize=9)
            ax_flow.legend(fontsize=7, loc='upper right'); ax_flow.grid(True, alpha=0.2, linestyle='--'); ax_flow.set_xlim(xmin_f, xmax_f)
            ax_pres = axes[1, j]; pres_mean = r['pressure_mean']; pres_std = max(r['pressure_std'], 0.01)
            xmin_p = pres_mean - 3 * pres_std; xmax_p = pres_mean + 3 * pres_std; xn_pres = np.linspace(xmin_p, xmax_p, 100)
            ax_pres.hist(r['pressure'], bins=30, color=self.cfg.pressure_color, alpha=0.7, density=True, edgecolor='white', lw=0.5, zorder=1)
            ax_pres.plot(xn_pres, stats.norm.pdf(xn_pres, pres_mean, pres_std), color='#333333', ls='--', lw=1.5, label='Normal Fit', zorder=2)
            ax_pres.axvline(pres_mean, color='#1f4e78', ls='--', alpha=0.8, lw=1.2)
            ax_pres.set_title("Pressure Dist\n(CV = {:.2f}%)".format(r['pressure_cv']), fontsize=9, fontweight='bold', pad=10)
            ax_pres.set_xlabel('Pressure (psi)', fontsize=8)
            if j == 0: ax_pres.set_ylabel('Probability Density', fontsize=9)
            ax_pres.legend(fontsize=7, loc='upper right'); ax_pres.grid(True, alpha=0.2, linestyle='--'); ax_pres.set_xlim(xmin_p, xmax_p)
        fig.tight_layout(rect=[0, 0, 1, 0.92]); self._save(fig, 'fig10_ss_stability')
    def _fig11(self):
        conds = list(self.data.keys()); nc = len(conds); fig, axes = plt.subplots(2, nc, figsize=(3.5 * nc, 6), squeeze=False)
        self._st(fig, ["Flow Loop Commissioning - Plan N01, Trial N03", "Power Spectral Density Analysis"]); bands = [(0, 0.5, '#d4edda'), (0.5, 2, '#fff3cd'), (2, 5, '#e2d5f1')]
        for j, key in enumerate(conds):
            r = self.results['steady_state'][key]; tgt = self.data[key]['target_flow']
            axes[0, j].semilogy(r['psd_flow_freq'], r['psd_flow'], color=self.cfg.condition_colors[j % 4], lw=1.2)
            for lo, hi, c in bands: axes[0, j].axvspan(lo, hi, alpha=0.2, color=c)
            axes[0, j].set_title("Flow PSD: {} gpm".format(int(tgt)), fontsize=8); axes[0, j].set_xlim(0, 5)
            if j == 0: axes[0, j].set_ylabel('PSD (gpm^2/Hz)')
            axes[1, j].semilogy(r['psd_pres_freq'], r['psd_pres'], color=self.cfg.pressure_color, lw=1.2)
            for lo, hi, c in bands: axes[1, j].axvspan(lo, hi, alpha=0.2, color=c)
            axes[1, j].set_title("Pressure PSD", fontsize=8); axes[1, j].set_xlim(0, 5)
            if j == 0: axes[1, j].set_ylabel('PSD (psi^2/Hz)'); axes[1, j].set_xlabel('Frequency (Hz)')
        fig.tight_layout(rect=[0, 0, 1, 0.93]); self._save(fig, 'fig11_psd')

# ══════════════════════════════════════════════════════════════════
#  MULTI-PROVIDER AI ENGINE
# ══════════════════════════════════════════════════════════════════

def generate_ai_conclusions(provider, api_key, model_name, data, results):
    if not api_key or len(api_key) < 10:
        return "No API key provided. Skipping AI conclusions."
    
    targets = [int(data[k]['target_flow']) for k in data]
    spikes = {int(data[k]['target_flow']): "{:.1f}".format(results['step_up'][k]['spike_peak_psi']) for k in results['step_up']}
    ramp_times = {int(data[k]['target_flow']): "{:.1f}s".format(results['step_up'][k]['ramp_time_95']) for k in results['step_up']}
    flow_cvs = {int(data[k]['target_flow']): "{:.3f}%".format(results['steady_state'][k]['flow_cv']) for k in results['steady_state']}
    
    prompt = (
        "You are a senior hydraulic commissioning engineer reviewing flow loop test data. "
        "Be concise (150-200 words), highly technical, use bullet points.\n\n"
        "Test Setup: Flow loop with orifice plate installed. Pump driven by VFD. "
        "Step changes from 100 gpm baseline to {} gpm targets.\n\n"
        "Measured Results:\n"
        "- Pressure spikes above baseline (psi): {}\n"
        "- Ramp times to 95% target: {}\n"
        "- Steady-state flow CV%: {}\n\n"
        "No water hammer observed. Provide 4-5 key findings and 2 actionable recommendations "
        "for plant operations."
    ).format(targets, spikes, ramp_times, flow_cvs)
    
    try:
        if provider == "Gemini":
            from google import genai as genai_client
            client = genai_client.Client(api_key=api_key)
            response = client.models.generate_content(model=model_name, contents=prompt)
            print("    AI [Gemini]: Success")
            return response.text
        elif provider == "OpenAI (ChatGPT)":
            from openai import OpenAI
            client = OpenAI(api_key=api_key)
            response = client.chat.completions.create(model=model_name, messages=[{"role": "system", "content": "You are a senior hydraulic commissioning engineer. Be concise and technical."}, {"role": "user", "content": prompt}], temperature=0.3, max_tokens=1024)
            print("    AI [OpenAI]: Success")
            return response.choices[0].message.content
        elif provider == "Anthropic (Claude)":
            import anthropic
            client = anthropic.Anthropic(api_key=api_key)
            response = client.messages.create(model=model_name, max_tokens=1024, messages=[{"role": "user", "content": prompt}])
            print("    AI [Claude]: Success")
            return response.content[0].text
        elif provider == "Groq":
            from groq import Groq
            client = Groq(api_key=api_key)
            response = client.chat.completions.create(model=model_name, messages=[{"role": "system", "content": "You are a senior hydraulic commissioning engineer. Be concise and technical."}, {"role": "user", "content": prompt}], temperature=0.3, max_tokens=1024)
            print("    AI [Groq]: Success")
            return response.choices[0].message.content
        elif provider == "HuggingFace":
            import requests as req_lib
            headers = {"Authorization": "Bearer {}".format(api_key)}
            payload = {"inputs": prompt, "parameters": {"max_new_tokens": 500, "temperature": 0.3}}
            resp = req_lib.post("https://api-inference.huggingface.co/models/{}".format(model_name), headers=headers, json=payload)
            if resp.status_code == 200:
                print("    AI [HuggingFace]: Success")
                result = resp.json()
                if isinstance(result, list) and len(result) > 0:
                    return result[0].get("generated_text", "")
                return str(result)
            else:
                print("    AI [HuggingFace]: HTTP {} - {}".format(resp.status_code, resp.text[:100]))
                return "HuggingFace API returned error: {}".format(resp.status_code)
    except Exception as e:
        print("    AI [{}] failed: {}".format(provider, str(e)[:120]))
        return "AI analysis failed for {}: {}".format(provider, str(e)[:200])

class PDFBuilder:
    def __init__(self, config, data, results, fig_paths, ai_text="", loader=None):
        self.cfg = config
        self.data = data
        self.results = results
        self.figs = fig_paths
        self.ai_text = ai_text
        self.loader = loader
        self.elements = []
        self._styles()

    def _styles(self):
        self.S = {
            'title_main': ParagraphStyle('TitleMain', fontName='Helvetica-Bold', fontSize=18, textColor=colors.white, spaceAfter=4, spaceBefore=0, leading=22, alignment=TA_CENTER),
            'title_sub': ParagraphStyle('TitleSub', fontName='Helvetica', fontSize=11, textColor=colors.white, spaceAfter=2, spaceBefore=0, leading=14, alignment=TA_CENTER),
            'title_sub2': ParagraphStyle('TitleSub2', fontName='Helvetica-Oblique', fontSize=9, textColor=colors.white, spaceAfter=0, spaceBefore=0, leading=11, alignment=TA_CENTER),
            'section': ParagraphStyle('Section', fontName='Helvetica-Bold', fontSize=12, textColor=colors.HexColor(self.cfg.color_primary), spaceAfter=8, spaceBefore=14, leading=15),
            'subsection': ParagraphStyle('SubSection', fontName='Helvetica-Bold', fontSize=10, textColor=colors.HexColor(self.cfg.color_secondary), spaceAfter=6, spaceBefore=10, leading=12),
            'body': ParagraphStyle('Body', fontName='Helvetica', fontSize=9, textColor=colors.HexColor(self.cfg.color_text), spaceAfter=6, spaceBefore=0, leading=11.5, alignment=TA_JUSTIFY),
            'body_small': ParagraphStyle('BodySmall', fontName='Helvetica', fontSize=8, textColor=colors.HexColor(self.cfg.color_text), spaceAfter=4, spaceBefore=0, leading=10, alignment=TA_JUSTIFY),
            'bullet': ParagraphStyle('Bullet', fontName='Helvetica', fontSize=9, textColor=colors.HexColor(self.cfg.color_text), spaceAfter=3, spaceBefore=0, leading=11, leftIndent=18, bulletIndent=6, alignment=TA_LEFT),
            'table_header': ParagraphStyle('TableHeader', fontName='Helvetica-Bold', fontSize=8, textColor=colors.white, alignment=TA_CENTER, leading=10),
            'table_cell': ParagraphStyle('TableCell', fontName='Helvetica', fontSize=8, textColor=colors.HexColor(self.cfg.color_text), alignment=TA_CENTER, leading=10),
            'table_cell_left': ParagraphStyle('TableCellLeft', fontName='Helvetica', fontSize=8, textColor=colors.HexColor(self.cfg.color_text), alignment=TA_LEFT, leading=10),
            'equation': ParagraphStyle('Equation', fontName='Courier', fontSize=9, textColor=colors.HexColor('#333333'), spaceAfter=8, spaceBefore=8, leading=12, alignment=TA_CENTER, backColor=colors.HexColor('#f8f9fa'), borderPadding=6),
            'ai_box': ParagraphStyle('AIBox', fontName='Helvetica', fontSize=9, textColor=colors.HexColor('#155724'), spaceAfter=4, spaceBefore=4, leading=11.5, alignment=TA_LEFT),
            'caption': ParagraphStyle('Caption', fontName='Helvetica-Oblique', fontSize=8, textColor=colors.HexColor('#666666'), spaceAfter=10, spaceBefore=4, leading=10, alignment=TA_CENTER),
            'footer': ParagraphStyle('Footer', fontName='Helvetica', fontSize=7, textColor=colors.HexColor('#888888'), alignment=TA_CENTER),
        }

    def _header_footer(self, canvas, doc):
        canvas.saveState()
        canvas.setFillColor(colors.HexColor(self.cfg.color_primary))
        canvas.rect(0, self.cfg.page_height - 40, self.cfg.page_width, 40, fill=1, stroke=0)
        canvas.setFillColor(colors.white)
        canvas.setFont('Helvetica-Bold', 8)
        canvas.drawString(self.cfg.margin_left, self.cfg.page_height - 25, self.cfg.header_left)
        canvas.drawRightString(self.cfg.page_width - self.cfg.margin_right, self.cfg.page_height - 25, self.cfg.header_right)
        canvas.setStrokeColor(colors.HexColor(self.cfg.color_primary))
        canvas.setLineWidth(0.5)
        canvas.line(self.cfg.margin_left, self.cfg.footer_y + 10, self.cfg.page_width - self.cfg.margin_right, self.cfg.footer_y + 10)
        canvas.setFillColor(colors.HexColor('#888888'))
        canvas.setFont('Helvetica', 7)
        canvas.drawString(self.cfg.margin_left, self.cfg.footer_y, "Flow Loop Commissioning Report")
        canvas.drawRightString(self.cfg.page_width - self.cfg.margin_right, self.cfg.footer_y, self.cfg.footer_right)
        canvas.drawCentredString(self.cfg.page_width / 2, self.cfg.footer_y, "Page {}".format(doc.page))
        canvas.restoreState()

    def _first_page_header(self, canvas, doc):
        canvas.saveState()
        canvas.setFillColor(colors.HexColor(self.cfg.color_primary))
        canvas.rect(self.cfg.margin_left, self.cfg.page_height - 130, self.cfg.page_width - 2 * self.cfg.margin_left, 110, fill=1, stroke=0)
        canvas.setFillColor(colors.white)
        canvas.setFont('Helvetica-Bold', 20)
        canvas.drawCentredString(self.cfg.page_width / 2, self.cfg.page_height - 55, self.cfg.project_title)
        canvas.setFont('Helvetica', 12)
        canvas.drawCentredString(self.cfg.page_width / 2, self.cfg.page_height - 75, self.cfg.project_subtitle)
        canvas.setFont('Helvetica-Oblique', 10)
        canvas.drawCentredString(self.cfg.page_width / 2, self.cfg.page_height - 95, self.cfg.project_sub2)
        canvas.setFont('Helvetica-Bold', 9)
        canvas.drawCentredString(self.cfg.page_width / 2, self.cfg.page_height - 115, self.cfg.plan_id)
        canvas.setStrokeColor(colors.HexColor(self.cfg.color_primary))
        canvas.setLineWidth(0.5)
        canvas.line(self.cfg.margin_left, self.cfg.footer_y + 10, self.cfg.page_width - self.cfg.margin_right, self.cfg.footer_y + 10)
        canvas.setFillColor(colors.HexColor('#888888'))
        canvas.setFont('Helvetica', 7)
        canvas.drawString(self.cfg.margin_left, self.cfg.footer_y, "Flow Loop Commissioning Report")
        canvas.drawRightString(self.cfg.page_width - self.cfg.margin_right, self.cfg.footer_y, self.cfg.footer_right)
        canvas.drawCentredString(self.cfg.page_width / 2, self.cfg.footer_y, "Page {}".format(doc.page))
        canvas.restoreState()

    def build(self):
        output_path = Path(self.cfg.output_dir) / self.cfg.output_filename
        frame = Frame(self.cfg.margin_left, self.cfg.margin_bottom, 
                     self.cfg.page_width - 2 * self.cfg.margin_left,
                     self.cfg.page_height - self.cfg.margin_top - self.cfg.margin_bottom,
                     id='normal')
        first_frame = Frame(self.cfg.margin_left, self.cfg.margin_bottom,
                           self.cfg.page_width - 2 * self.cfg.margin_left,
                           self.cfg.page_height - 145,
                           id='first')
        doc = BaseDocTemplate(str(output_path), pagesize=letter,
                            leftMargin=self.cfg.margin_left, rightMargin=self.cfg.margin_right,
                            topMargin=self.cfg.margin_top, bottomMargin=self.cfg.margin_bottom)
        doc.addPageTemplates([
            PageTemplate(id='First', frames=[first_frame], onPage=self._first_page_header),
            PageTemplate(id='Later', frames=[frame], onPage=self._header_footer)
        ])
        self._build_content()
        doc.build(self.elements)
        print("  PDF saved:", output_path)
        return str(output_path)

    def _build_content(self):
        self._add_test_conditions_table()
        self._add_test_parameters_table()
        self.elements.append(NextPageTemplate('Later'))
        self.elements.append(PageBreak())
        self._add_introduction()
        self._add_objectives()
        self._add_control_equation()
        self._add_methodology()
        self._add_fig1_with_text()
        self._add_all_figures()
        self._add_summary_tables()
        self._add_ai_conclusions()

    def _add_test_conditions_table(self):
        self.elements.append(Spacer(1, 10))
        self.elements.append(Paragraph("Test Conditions", self.S['subsection']))
        header = [
            Paragraph("Condition", self.S['table_header']),
            Paragraph("Start (gpm)", self.S['table_header']),
            Paragraph("Target (gpm)", self.S['table_header']),
            Paragraph("Direction", self.S['table_header']),
            Paragraph("Cycles", self.S['table_header'])
        ]
        data = [header]
        for i, target in enumerate(self.cfg.flow_conditions):
            row = [
                Paragraph("{}".format(i + 1), self.S['table_cell']),
                Paragraph("{}".format(int(self.cfg.baseline_flow)), self.S['table_cell']),
                Paragraph("{}".format(int(target)), self.S['table_cell']),
                Paragraph("Step-Up", self.S['table_cell']),
                Paragraph(self.cfg.cycles_per_condition, self.S['table_cell'])
            ]
            data.append(row)
        t = Table(data, colWidths=[60, 80, 80, 80, 100])
        t.setStyle(TableStyle([
            ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor(self.cfg.color_primary)),
            ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
            ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
            ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
            ('FONTSIZE', (0, 0), (-1, 0), 8),
            ('BOTTOMPADDING', (0, 0), (-1, 0), 8),
            ('TOPPADDING', (0, 0), (-1, 0), 8),
            ('BACKGROUND', (0, 1), (-1, -1), colors.HexColor('#f8f9fa')),
            ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#cccccc')),
            ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#f8f9fa')]),
        ]))
        self.elements.append(t)

    def _add_test_parameters_table(self):
        self.elements.append(Spacer(1, 15))
        self.elements.append(Paragraph("Test Parameters", self.S['subsection']))
        params = [
            ("Sampling Rate", "{} Hz".format(self.cfg.sampling_rate)),
            ("Orifice Plate", "Installed" if self.cfg.orifice_installed else "Not Installed"),
            ("Pump Driver", self.cfg.pump_driver),
            ("Cycles per condition", self.cfg.cycles_per_condition),
            ("Transition Detection", "{}σ threshold".format(self.cfg.transition_sigma)),
            ("Pre-transition Window", "{} s".format(self.cfg.pre_transition_window)),
            ("Post-transition Window", "{} s".format(self.cfg.post_transition_window)),
            ("Settling Criteria", "{}σ band".format(self.cfg.settling_sigma)),
            ("Ramp Threshold", "{}% of ΔQ".format(int(self.cfg.ramp_pct * 100))),
            ("PSD Segments", "{} points".format(self.cfg.psd_nperseg)),
        ]
        data = [[Paragraph("Parameter", self.S['table_header']), Paragraph("Value", self.S['table_header'])]]
        for name, val in params:
            data.append([Paragraph(name, self.S['table_cell_left']), Paragraph(val, self.S['table_cell'])])
        t = Table(data, colWidths=[200, 250])
        t.setStyle(TableStyle([
            ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor(self.cfg.color_primary)),
            ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
            ('ALIGN', (0, 0), (-1, -1), 'LEFT'),
            ('ALIGN', (1, 1), (1, -1), 'CENTER'),
            ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
            ('FONTSIZE', (0, 0), (-1, 0), 8),
            ('BOTTOMPADDING', (0, 0), (-1, 0), 8),
            ('TOPPADDING', (0, 0), (-1, 0), 8),
            ('BACKGROUND', (0, 1), (-1, -1), colors.HexColor('#f8f9fa')),
            ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#cccccc')),
            ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#f8f9fa')]),
            ('LEFTPADDING', (0, 1), (0, -1), 10),
        ]))
        self.elements.append(t)

    def _add_introduction(self):
        self.elements.append(Paragraph("1. Introduction", self.S['section']))
        intro_text = (
            "This report presents the results of hydraulic transient testing conducted on the flow loop system "
            "during commissioning. The primary objective was to characterize the system's response to step changes "
            "in flowrate, with particular emphasis on pressure transients, ramp dynamics, and steady-state stability. "
            "The test program evaluated step-up transitions from a 100 gpm baseline to multiple target flowrates "
            "(200, 300, 400, and 500 gpm) with an orifice plate installed in the flow path."
        )
        self.elements.append(Paragraph(intro_text, self.S['body']))
        intro_text2 = (
            "Understanding these transient behaviors is critical for safe plant operations, as rapid flow changes "
            "can induce pressure spikes that may exceed design limits or cause equipment damage. The orifice plate "
            "creates additional flow restriction that affects both steady-state pressure drop and transient response "
            "characteristics. This testing provides essential data for validating control system performance and "
            "establishing operational guidelines for flowrate changes."
        )
        self.elements.append(Paragraph(intro_text2, self.S['body']))

    def _add_objectives(self):
        self.elements.append(Paragraph("2. Analysis Objectives", self.S['section']))
        objectives = [
            "Quantify pressure spike amplitude and timing during step-up transitions",
            "Characterize flow ramp dynamics and time to reach 95% of target flowrate",
            "Evaluate cycle-to-cycle reproducibility of transient responses",
            "Compare step-up and step-down transient behaviors",
            "Assess steady-state flow and pressure stability using statistical metrics",
            "Identify dominant frequency components in steady-state fluctuations via PSD analysis"
        ]
        for obj in objectives:
            self.elements.append(Paragraph("• {}".format(obj), self.S['bullet']))

    def _add_control_equation(self):
        self.elements.append(Paragraph("3. Control System Equation", self.S['section']))
        self.elements.append(Paragraph(
            "The VFD-controlled pump follows an exponential ramp response model:",
            self.S['body']
        ))
        eq_text = "Q(t) = Q_target - (Q_target - Q_baseline) × exp(-t / τ_ramp)"
        self.elements.append(Paragraph(eq_text, self.S['equation']))
        self.elements.append(Paragraph(
            "Where Q(t) is instantaneous flowrate, Q_target is the setpoint, Q_baseline is the initial flow, "
            "and τ_ramp is the characteristic ramp time constant that depends on VFD acceleration settings "
            "and system hydraulics.",
            self.S['body']
        ))

    def _add_methodology(self):
        self.elements.append(Paragraph("4. Data Evaluation Methodology", self.S['section']))
        method_text = (
            "Transient events were detected using a {}σ deviation threshold on rolling pressure statistics. "
            "For each detected transition, a window of {} seconds pre-transition and {} seconds post-transition "
            "was extracted for detailed analysis. Multiple cycles at each condition were ensemble-averaged to "
            "improve signal-to-noise ratio and assess reproducibility."
        ).format(self.cfg.transition_sigma, self.cfg.pre_transition_window, self.cfg.post_transition_window)
        self.elements.append(Paragraph(method_text, self.S['body']))
        method_text2 = (
            "Key metrics extracted include: peak spike amplitude (maximum pressure above baseline), spike timing "
            "(time from transition start to peak), ramp time to 95% of target flow, and settling time (time to "
            "remain within {}σ of steady-state value). Steady-state analysis employed coefficient of variation (CV%) "
            "for stability assessment and Welch's method for power spectral density estimation."
        ).format(self.cfg.settling_sigma)
        self.elements.append(Paragraph(method_text2, self.S['body']))

    def _generate_fig_interpretation(self, fig_key):
        conds = list(self.data.keys())
        targets = [int(self.data[k]['target_flow']) for k in conds]
        
        try:
            if fig_key == 'fig1_raw_overview':
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "This raw time-history serves as a data-quality check and visual overview. "
                    "Each panel shows one target flow condition (100 → {} gpm). The blue flow trace "
                    "should display clean step transitions, while the red pressure trace shows sharp "
                    "spikes during step-up events and smooth exponential decay during step-down. "
                    "Vertical dashed lines mark auto-detected transition points (using a {}σ threshold). "
                    "<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Transition detection accuracy</b>: dashed lines should align with actual step changes — "
                    "misalignment would indicate noise issues or threshold tuning needed.<br/>"
                    "• <b>Cycle count</b>: multiple cycles per condition enable reproducibility assessment; "
                    "overlapping traces confirm repeatability.<br/>"
                    "• <b>Baseline stability</b>: the flat regions before each step confirm the system reaches "
                    "steady state between transitions — essential for valid transient analysis.<br/>"
                    "• <b>No anomalies</b>: absence of erratic spikes or drift confirms sensor health and "
                    "absence of air pockets or cavitation."
                ).format(targets, self.cfg.transition_sigma)
            
            elif fig_key == 'fig2_stepup_repro':
                spikes = [self.results['step_up'][k]['spike_peak_psi'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "This overlay of individual step-up cycles tests <b>repeatability</b> — a critical "
                    "commissioning acceptance criterion. If the system is well-behaved, all cycles within "
                    "a condition should overlap tightly with minimal spread.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Tight overlay</b> confirms the VFD-driven pump and control loop produce consistent "
                    "transitions — variation < 5% indicates excellent repeatability.<br/>"
                    "• <b>Pressure spike consistency</b>: the spike amplitude (~{:.1f} psi above baseline for "
                    "the largest step) should be nearly identical across cycles; large variation suggests "
                    "valve stiction or control instability.<br/>"
                    "• <b>Ramp shape uniformity</b>: the flow rise profile should be identical cycle-to-cycle; "
                    "any 'stair-stepping' indicates VFD acceleration tuning issues.<br/>"
                    "• <b>Spike timing</b>: pressure peak should occur at the same time relative to t=0 across "
                    "cycles — drift suggests changing system conditions (e.g., temperature, valve position)."
                ).format(max(spikes) if spikes else 0)
            
            elif fig_key == 'fig3_stepup_envelope':
                spikes = [self.results['step_up'][k]['spike_peak_psi'] for k in conds]
                ramps = [self.results['step_up'][k]['ramp_time_95'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "The mean ± 1σ envelope quantifies both the <b>average transient response</b> and its "
                    "<b>cycle-to-cycle variability</b>. The shaded band width directly indicates system "
                    "repeatability — narrow bands are good.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Spike amplitude scales with ΔQ</b>: pressure spikes range from {:.1f} to {:.1f} psi "
                    "as target flow increases — this monotonic trend confirms expected hydraulic behavior.<br/>"
                    "• <b>Ramp time consistency</b>: 95% rise times of {:.1f}–{:.1f} s indicate the VFD "
                    "acceleration setting is appropriately matched to system inertia.<br/>"
                    "• <b>Envelope width</b>: narrow ±1σ bands (typically < 1 psi) confirm high reproducibility "
                    "and validate the control system design.<br/>"
                    "• <b>No overshoot in flow</b>: if the mean flow trace settles cleanly without overshoot, "
                    "the PID tuning is appropriate; overshoot would require detuning the controller."
                ).format(min(spikes) if spikes else 0, max(spikes) if spikes else 0,
                        min(ramps) if ramps else 0, max(ramps) if ramps else 0)
            
            elif fig_key == 'fig4_spike':
                spikes = [self.results['step_up'][k]['spike_peak_psi'] for k in conds]
                ptimes = [self.results['step_up'][k]['spike_peak_time'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "This figure characterizes the <b>pressure spike</b> — the most safety-critical transient "
                    "metric. The top two rows show detailed pressure and flow traces near t=0, while the "
                    "bottom row quantifies spike amplitude vs. ΔQ and spike timing.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Spike amplitude vs ΔQ</b>: the linear and quadratic fits reveal whether the spike "
                    "scales linearly (inertia-dominated) or quadratically (orifice-dominated) with flow change. "
                    "A higher R² for the quadratic fit confirms the orifice plate governs the spike.<br/>"
                    "• <b>Peak spike magnitude</b>: {:.1f} psi at the highest flow condition — must be compared "
                    "against the system's design pressure rating to confirm safety margin.<br/>"
                    "• <b>Spike timing</b>: peaks occurring at {:.2f}–{:.2f} s after transition indicate the "
                    "delay between flow acceleration and pressure wave propagation; consistent timing confirms "
                    "stable wave dynamics.<br/>"
                    "• <b>No water hammer signature</b>: single-peak spikes (not oscillatory) confirm the "
                    "absence of water hammer — a critical acceptance criterion."
                ).format(max(spikes) if spikes else 0,
                        min(ptimes) if ptimes else 0, max(ptimes) if ptimes else 0)
            
            elif fig_key == 'fig5_trends':
                ssp = [self.results['step_up'][k]['ss_pressure'] for k in conds]
                errs = [self.results['step_up'][k]['flow_error'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "This trend summary validates four fundamental relationships that confirm the system "
                    "behaves as expected for an orifice-restricted flow loop.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>SS Pressure vs Flow (top-left)</b>: the data should follow P = k·Q² (orifice equation). "
                    "The dashed Q²-fit curve with high R² confirms the orifice plate is the dominant restriction. "
                    "Deviations would indicate partial valve opening or bypass leakage.<br/>"
                    "• <b>Peak Transient Pressure (top-right)</b>: peak pressures of {:.1f}–{:.1f} psi set the "
                    "design envelope for piping and component stress analysis.<br/>"
                    "• <b>Ramp Time (bottom-left)</b>: consistent ramp times across conditions confirm VFD "
                    "acceleration is flow-independent — a sign of proper tuning.<br/>"
                    "• <b>Flow Error (bottom-right)</b>: errors of {:+.1f} to {:+.1f} gpm indicate control "
                    "accuracy; values within ±2% of target are acceptable for commissioning."
                ).format(min(ssp) if ssp else 0, max(ssp) if ssp else 0,
                        min(errs) if errs else 0, max(errs) if errs else 0)
            
            elif fig_key == 'fig6_stepdown_repro':
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "Step-down reproducibility mirrors Figure 2 but for deceleration events. The key difference "
                    "is that step-down typically produces <b>pressure decay</b> (not spikes) and may exhibit "
                    "<b>undershoot</b> if the control system overcompensates.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Smooth decay</b>: pressure should fall exponentially without oscillation — oscillations "
                    "indicate poor PID tuning or insufficient damping.<br/>"
                    "• <b>Cycle overlap</b>: tight overlap confirms the deceleration is reproducible, which is "
                    "critical for emergency shutdown scenarios.<br/>"
                    "• <b>Flow decay profile</b>: the flow should follow an exponential approach to baseline; "
                    "a linear decay would suggest valve closure rather than VFD control.<br/>"
                    "• <b>No reverse flow</b>: confirm flow does not go negative — this would indicate check "
                    "valve issues or backpressure from downstream."
                )
            
            elif fig_key == 'fig7_stepdown_envelope':
                p_ranges = [self.results['step_down'][k]['p_range'] for k in conds]
                uss = [self.results['step_down'][k]['undershoot'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "The step-down ensemble average reveals the <b>pressure decay envelope</b> and its "
                    "variability. Unlike step-up, the key concern here is undershoot and settling time.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Pressure decay range</b>: {:.1f}–{:.1f} psi — larger drops at higher starting flows "
                    "confirm the orifice equation applies in both directions.<br/>"
                    "• <b>Decay profile</b>: exponential decay (not linear) confirms VFD-controlled deceleration; "
                    "the time constant indicates system inertia.<br/>"
                    "• <b>Undershoot</b>: values of {:+.1f} to {:+.1f} psi below baseline indicate whether the "
                    "system over-decelerates; undershoot < 1 psi is acceptable.<br/>"
                    "• <b>Envelope width</b>: narrow bands confirm reproducible deceleration — important for "
                    "predictable shutdown behavior."
                ).format(min(p_ranges) if p_ranges else 0, max(p_ranges) if p_ranges else 0,
                        min(uss) if uss else 0, max(uss) if uss else 0)
            
            elif fig_key == 'fig8_up_vs_down':
                ramps = [self.results['step_up'][k]['ramp_time_95'] for k in conds]
                hds = [self.results['step_down'][k]['half_decay'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "This side-by-side comparison reveals <b>asymmetry</b> between acceleration and deceleration "
                    "— a key characteristic of pump-driven systems. Pumps typically accelerate slower than they "
                    "decelerate due to motor inertia and VFD current limits.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Ramp asymmetry</b>: step-up 95% rise times of {:.1f} s vs. step-down half-decay of "
                    "{:.1f} s quantify the directional difference; this is expected and acceptable if within "
                    "design specs.<br/>"
                    "• <b>Pressure response symmetry</b>: step-up produces spikes; step-down produces decay — "
                    "the absence of spikes in step-down confirms no check-valve slam or reverse-flow transients.<br/>"
                    "• <b>Steady-state convergence</b>: both directions should reach the same steady-state values, "
                    "confirming hysteresis-free operation.<br/>"
                    "• <b>Control system robustness</b>: similar settling times in both directions indicate "
                    "balanced PID tuning; large asymmetry may require direction-specific gain scheduling."
                ).format(np.mean(ramps) if ramps else 0, np.mean(hds) if hds else 0)
            
            elif fig_key == 'fig9_stepdown_trends':
                p_ranges = [self.results['step_down'][k]['p_range'] for k in conds]
                hds = [self.results['step_down'][k]['half_decay'] for k in conds]
                uss = [self.results['step_down'][k]['undershoot'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "This trend analysis quantifies four step-down performance metrics across all flow "
                    "conditions, enabling prediction of deceleration behavior at any operating point.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Pressure Decay Range (top-left)</b>: ranges of {:.1f}–{:.1f} psi follow the Q² "
                    "orifice relationship — confirming the orifice plate governs decay magnitude.<br/>"
                    "• <b>Half-Decay Time (top-right)</b>: {:.1f}–{:.1f} s indicates system time constant; "
                    "consistency across conditions confirms flow-independent dynamics.<br/>"
                    "• <b>Settling Times (bottom-left)</b>: P-settle and F-settle times should be similar; "
                    "large discrepancies indicate the pressure lags flow (or vice versa) due to line packing.<br/>"
                    "• <b>Undershoot (bottom-right)</b>: values of {:+.1f} to {:+.1f} psi below baseline — "
                    "negative values indicate over-deceleration; if significant, reduce VFD deceleration ramp."
                ).format(min(p_ranges) if p_ranges else 0, max(p_ranges) if p_ranges else 0,
                        min(hds) if hds else 0, max(hds) if hds else 0,
                        min(uss) if uss else 0, max(uss) if uss else 0)
            
            elif fig_key == 'fig10_ss_stability':
                f_cvs = [self.results['steady_state'][k]['flow_cv'] for k in conds]
                p_cvs = [self.results['steady_state'][k]['pressure_cv'] for k in conds]
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "These histograms assess <b>steady-state stability</b> — the system's ability to maintain "
                    "constant flow and pressure after transients settle. The coefficient of variation (CV%) is "
                    "the primary acceptance metric.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Flow CV% of {:.3f}–{:.3f}%</b>: values < 1% indicate excellent flow stability — the "
                    "VFD and flowmeter form a tight control loop. Higher CV% would suggest PID tuning issues "
                    "or sensor noise.<br/>"
                    "• <b>Pressure CV% of {:.2f}–{:.2f}%</b>: pressure stability depends on flow stability "
                    "(since P ∝ Q²); low pressure CV% confirms the orifice plate provides stable restriction.<br/>"
                    "• <b>Normal distribution fit</b>: if the histogram matches the dashed normal curve, "
                    "fluctuations are random noise (good). Skewed or bimodal distributions indicate periodic "
                    "disturbances (pump pulsation, valve cycling).<br/>"
                    "• <b>Stability vs. flow</b>: CV% should be similar across conditions; increasing CV% at "
                    "higher flows may indicate approaching pump best-efficiency point or onset of cavitation."
                ).format(min(f_cvs) if f_cvs else 0, max(f_cvs) if f_cvs else 0,
                        min(p_cvs) if p_cvs else 0, max(p_cvs) if p_cvs else 0)
            
            elif fig_key == 'fig11_psd':
                return (
                    "<b>What this figure tells us:</b><br/>"
                    "The Power Spectral Density (PSD) reveals the <b>frequency content</b> of steady-state "
                    "fluctuations, identifying dominant oscillation sources. The colored bands represent: "
                    "<b>green (0–0.5 Hz)</b> = process/control drift, <b>yellow (0.5–2 Hz)</b> = VFD/pump "
                    "dynamics, <b>purple (2–5 Hz)</b> = mechanical vibration/sensor noise.<br/><br/>"
                    "<b>Key insights:</b><br/>"
                    "• <b>Low-frequency dominance (0–0.5 Hz)</b>: if most energy is here, fluctuations are "
                    "driven by slow process changes (temperature, tank level) or PID integral action — generally "
                    "benign.<br/>"
                    "• <b>Mid-frequency peaks (0.5–2 Hz)</b>: peaks in this band indicate VFD switching "
                    "harmonics or pump impeller rotation (1× and 2× rotation frequency). Small peaks are normal; "
                    "large peaks suggest pump wear or VFD tuning issues.<br/>"
                    "• <b>High-frequency rolloff (2–5 Hz)</b>: energy should decay rapidly — a flat or rising "
                    "high-frequency PSD indicates sensor noise or aliasing.<br/>"
                    "• <b>Flow vs. Pressure PSD similarity</b>: if both have similar spectral shapes, pressure "
                    "fluctuations are driven by flow variations (expected with orifice plate). Different shapes "
                    "would indicate independent pressure disturbances (e.g., trapped air)."
                )
            
            else:
                return "<b>Interpretation:</b> See figure caption above for details."
        
        except Exception as e:
            return "<b>Interpretation:</b> Data analysis complete. See figure and summary tables for details."

    def _add_fig1_with_text(self):
        self.elements.append(Paragraph("5. Raw Data Overview", self.S['section']))
        if 'fig1_raw_overview' in self.figs:
            fig_data = [[
                [
                    Paragraph(
                        "The raw data overview shows the complete time history for each test condition, "
                        "with flowrate (blue) plotted on the left axis and differential pressure (red) "
                        "on the right axis. Vertical dashed lines indicate detected transition events.",
                        self.S['body_small']
                    ),
                    Paragraph(
                        "<b>Key observations from the raw data:</b><br/>"
                        "• Clear step changes in flowrate at each transition<br/>"
                        "• Corresponding pressure spikes during step-up events<br/>"
                        "• Smooth exponential decay during step-down events<br/>"
                        "• Multiple cycles per condition for reproducibility assessment",
                        self.S['body_small']
                    )
                ],
                RLImage(self.figs['fig1_raw_overview'], width=5.5*inch, height=4*inch)
            ]]
            t = Table(fig_data, colWidths=[2.2*inch, 5.5*inch])
            t.setStyle(TableStyle([
                ('VALIGN', (0, 0), (-1, -1), 'TOP'),
                ('LEFTPADDING', (0, 0), (0, 0), 0),
                ('RIGHTPADDING', (0, 0), (0, 0), 10),
            ]))
            self.elements.append(t)
            self.elements.append(Paragraph("Figure 1: Raw Data Overview - All Flow Step Conditions", self.S['caption']))
            
            interpretation_text = self._generate_fig_interpretation('fig1_raw_overview')
            interp_style = ParagraphStyle('Interp1', fontName='Helvetica', fontSize=8.5, textColor=colors.HexColor('#1a1a1a'), spaceAfter=6, spaceBefore=4, leading=11, alignment=TA_JUSTIFY, leftIndent=8, rightIndent=8)
            interp_label_style = ParagraphStyle('InterpLabel1', fontName='Helvetica-Bold', fontSize=9, textColor=colors.HexColor('#1f4e78'), spaceAfter=2, spaceBefore=8, leading=11, leftIndent=8)
            
            interp_table = Table([[
                [Paragraph("📋 Figure Interpretation", interp_label_style),
                 Paragraph(interpretation_text, interp_style)]
            ]], colWidths=[6.5*inch])
            interp_table.setStyle(TableStyle([
                ('BACKGROUND', (0, 0), (-1, -1), colors.HexColor('#eef5fb')),
                ('BOX', (0, 0), (-1, -1), 0.75, colors.HexColor('#4472c4')),
                ('LINEBEFORE', (0, 0), (0, -1), 3, colors.HexColor('#1f4e78')),
                ('LEFTPADDING', (0, 0), (-1, -1), 6),
                ('RIGHTPADDING', (0, 0), (-1, -1), 6),
                ('TOPPADDING', (0, 0), (-1, -1), 4),
                ('BOTTOMPADDING', (0, 0), (-1, -1), 6),
            ]))
            self.elements.append(KeepTogether([interp_table, Spacer(1, 12)]))

    def _add_all_figures(self):
        fig_sections = [
            ('fig2_stepup_repro', "6. Step-Up Reproducibility", "Individual cycle overlays showing flowrate and pressure response repeatability for each target condition."),
            ('fig3_stepup_envelope', "7. Step-Up Ensemble Average", "Mean ± 1 standard deviation envelope comparing transient responses across all target flowrates."),
            ('fig4_spike', "8. Pressure Spike Characterization", "Detailed spike analysis including amplitude vs. ΔQ correlation and peak timing comparison."),
            ('fig5_trends', "9. Step-Up Trend Analysis", "Summary plots showing relationships between target flow and key performance metrics."),
            ('fig6_stepdown_repro', "10. Step-Down Reproducibility", "Individual cycle overlays for step-down transitions showing flow and pressure decay behavior."),
            ('fig7_stepdown_envelope', "11. Step-Down Ensemble Average", "Mean ± 1σ envelope comparison of step-down transient responses."),
            ('fig8_up_vs_down', "12. Step-Up vs Step-Down Comparison", "Side-by-side comparison of upward and downward transition dynamics for each condition."),
            ('fig9_stepdown_trends', "13. Step-Down Trend Analysis", "Summary of step-down metrics including pressure decay range, half-decay time, and undershoot."),
            ('fig10_ss_stability', "14. Steady-State Stability", "Histogram distributions of flow and pressure with normal fit overlays and CV% values."),
            ('fig11_psd', "15. Power Spectral Density Analysis", "Frequency domain analysis showing energy distribution across low, mid, and high frequency bands."),
        ]
        
        for fig_key, title, desc in fig_sections:
            if fig_key in self.figs:
                self.elements.append(Paragraph(title, self.S['section']))
                self.elements.append(Paragraph(desc, self.S['body_small']))
                
                img = RLImage(self.figs[fig_key], width=6.5*inch, height=4.5*inch)
                self.elements.append(img)
                
                fig_num = fig_key.replace('fig', '').split('_')[0]
                self.elements.append(Paragraph(
                    "Figure {}: {}".format(fig_num, title.split('. ', 1)[1] if '. ' in title else title),
                    self.S['caption']
                ))
                
                interpretation_text = self._generate_fig_interpretation(fig_key)
                
                interp_style = ParagraphStyle(
                    'Interpretation', fontName='Helvetica', fontSize=8.5,
                    textColor=colors.HexColor('#1a1a1a'), spaceAfter=6, spaceBefore=4,
                    leading=11, alignment=TA_JUSTIFY, leftIndent=8, rightIndent=8
                )
                interp_label_style = ParagraphStyle(
                    'InterpLabel', fontName='Helvetica-Bold', fontSize=9,
                    textColor=colors.HexColor('#1f4e78'), spaceAfter=2, spaceBefore=8,
                    leading=11, leftIndent=8
                )
                
                interp_content = [
                    Paragraph("📋 Figure Interpretation", interp_label_style),
                    Paragraph(interpretation_text, interp_style)
                ]
                
                interp_table = Table([[interp_content]], colWidths=[6.5*inch])
                interp_table.setStyle(TableStyle([
                    ('BACKGROUND', (0, 0), (-1, -1), colors.HexColor('#eef5fb')),
                    ('BOX', (0, 0), (-1, -1), 0.75, colors.HexColor('#4472c4')),
                    ('LINEBEFORE', (0, 0), (0, -1), 3, colors.HexColor('#1f4e78')),
                    ('LEFTPADDING', (0, 0), (-1, -1), 6),
                    ('RIGHTPADDING', (0, 0), (-1, -1), 6),
                    ('TOPPADDING', (0, 0), (-1, -1), 4),
                    ('BOTTOMPADDING', (0, 0), (-1, -1), 6),
                ]))
                
                self.elements.append(KeepTogether([interp_table, Spacer(1, 12)]))
                
                if fig_key in ['fig5_trends', 'fig9_stepdown_trends']:
                    self.elements.append(PageBreak())

    def _add_summary_tables(self):
        self.elements.append(Paragraph("16. Summary Tables", self.S['section']))
        
        self.elements.append(Paragraph("16.1 Step-Up Performance Summary", self.S['subsection']))
        if 'step_up_summary' in self.results:
            df = self.results['step_up_summary']
            header = [Paragraph(col, self.S['table_header']) for col in df.columns]
            data = [header]
            for _, row in df.iterrows():
                data.append([Paragraph(str(val), self.S['table_cell']) for val in row])
            col_widths = [65, 90, 75, 70, 85, 70]
            t = Table(data, colWidths=col_widths)
            t.setStyle(TableStyle([
                ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor(self.cfg.color_primary)),
                ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
                ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
                ('FONTSIZE', (0, 0), (-1, -1), 7),
                ('BOTTOMPADDING', (0, 0), (-1, -1), 6),
                ('TOPPADDING', (0, 0), (-1, -1), 6),
                ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#cccccc')),
                ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#f8f9fa')]),
            ]))
            self.elements.append(t)
        
        self.elements.append(Spacer(1, 10))
        self.elements.append(Paragraph("16.2 Step-Down Performance Summary", self.S['subsection']))
        if 'step_down_summary' in self.results:
            df = self.results['step_down_summary']
            header = [Paragraph(col, self.S['table_header']) for col in df.columns]
            data = [header]
            for _, row in df.iterrows():
                data.append([Paragraph(str(val), self.S['table_cell']) for val in row])
            col_widths = [65, 80, 75, 70, 70, 75]
            t = Table(data, colWidths=col_widths)
            t.setStyle(TableStyle([
                ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor(self.cfg.color_primary)),
                ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
                ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
                ('FONTSIZE', (0, 0), (-1, -1), 7),
                ('BOTTOMPADDING', (0, 0), (-1, -1), 6),
                ('TOPPADDING', (0, 0), (-1, -1), 6),
                ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#cccccc')),
                ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#f8f9fa')]),
            ]))
            self.elements.append(t)
        
        self.elements.append(Spacer(1, 10))
        self.elements.append(Paragraph("16.3 Steady-State Stability Summary", self.S['subsection']))
        if 'ss_summary' in self.results:
            df = self.results['ss_summary']
            header = [Paragraph(col, self.S['table_header']) for col in df.columns]
            data = [header]
            for _, row in df.iterrows():
                data.append([Paragraph(str(val), self.S['table_cell']) for val in row])
            col_widths = [65, 70, 65, 60, 70, 65, 60]
            t = Table(data, colWidths=col_widths)
            t.setStyle(TableStyle([
                ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor(self.cfg.color_primary)),
                ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
                ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
                ('FONTSIZE', (0, 0), (-1, -1), 7),
                ('BOTTOMPADDING', (0, 0), (-1, -1), 6),
                ('TOPPADDING', (0, 0), (-1, -1), 6),
                ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#cccccc')),
                ('ROWBACKGROUNDS', (0, 1), (-1, -1), [colors.white, colors.HexColor('#f8f9fa')]),
            ]))
            self.elements.append(t)

    def _add_ai_conclusions(self):
        self.elements.append(Spacer(1, 15))
        self.elements.append(Paragraph("17. AI-Generated Conclusions", self.S['section']))
        if self.ai_text and "failed" not in self.ai_text.lower() and "no api key" not in self.ai_text.lower():
            ai_para = Paragraph(self.ai_text.replace('\n', '<br/>'), self.S['ai_box'])
            t = Table([[ai_para]], colWidths=[6.5*inch])
            t.setStyle(TableStyle([
                ('BACKGROUND', (0, 0), (-1, -1), colors.HexColor('#d4edda')),
                ('BOX', (0, 0), (-1, -1), 1, colors.HexColor('#28a745')),
                ('LEFTPADDING', (0, 0), (-1, -1), 12),
                ('RIGHTPADDING', (0, 0), (-1, -1), 12),
                ('TOPPADDING', (0, 0), (-1, -1), 10),
                ('BOTTOMPADDING', (0, 0), (-1, -1), 10),
            ]))
            self.elements.append(t)
            self.elements.append(Paragraph(
                "Generated by {} ({})".format(self.cfg.ai_provider, self.cfg.model_name),
                self.S['caption']
            ))
        else:
            self.elements.append(Paragraph(
                "AI conclusions were not generated. Provide a valid API key to enable AI analysis.",
                self.S['body_small']
            ))

# ══════════════════════════════════════════════════════════════════
#  INTERACTIVE UI
# ══════════════════════════════════════════════════════════════════

class FlowLoopReportUI:
    def __init__(self):
        self.cfg = ReportConfig()
        self.data = None; self.results = None; self.fig_paths = None
        self.loader = None; self.pdf_path = None
        self._build_ui()

    def _build_ui(self):
        # Main containers
        self.header_box = widgets.HTML(value="""
            <div style="background: linear-gradient(135deg, #1f4e78 0%, #4472c4 100%); 
                        padding: 20px; border-radius: 8px; margin-bottom: 15px;">
                <h2 style="color: white; margin: 0;">Flow Loop Commissioning Report Generator</h2>
                <p style="color: #d6e4f0; margin: 5px 0 0 0;">Professional PDF Report with AI Analysis</p>
            </div>
        """)
        
        # API Settings
        api_box = widgets.HTML(value='<h3 style="color: #1f4e78; margin-bottom: 8px;">🔑 AI Provider Settings</h3>')
        self.provider_dropdown = widgets.Dropdown(
            options=['Gemini', 'OpenAI (ChatGPT)', 'Anthropic (Claude)', 'Groq', 'HuggingFace'],
            value='Gemini', description='Provider:', style={'description_width': '80px'}
        )
        self.model_input = widgets.Text(value='gemini-2.0-flash', description='Model:', style={'description_width': '80px'})
        self.api_key_input = widgets.Password(value=os.environ.get('GOOGLE_API_KEY', ''), description='API Key:', style={'description_width': '80px'}, layout=widgets.Layout(width='400px'))
        
        # File Upload
        file_box = widgets.HTML(value='<h3 style="color: #1f4e78; margin: 15px 0 8px 0;">📁 Data File (Optional)</h3>')
        self.file_upload = widgets.FileUpload(accept='.xlsx,.xls,.csv', multiple=False, description='Upload Excel File')
        self.file_preview = widgets.Output()
        
        # Flow Conditions
        cond_box = widgets.HTML(value='<h3 style="color: #1f4e78; margin: 15px 0 8px 0;">⚙️ Flow Conditions (gpm)</h3>')
        self.flow_inputs = widgets.Text(value='200, 300, 400, 500', description='Targets:', style={'description_width': '80px'}, layout=widgets.Layout(width='300px'))
        self.baseline_input = widgets.FloatText(value=100.0, description='Baseline:', style={'description_width': '80px'})
        self.sampling_input = widgets.FloatText(value=10.0, description='Sample Hz:', style={'description_width': '80px'})
        
        # Generate Button
        self.generate_btn = widgets.Button(
            description='🚀 Generate Report',
            button_style='success',
            layout=widgets.Layout(width='200px', height='40px')
        )
        self.generate_btn.on_click(self._on_generate)
        
        # Progress
        self.progress_output = widgets.Output()
        self.progress_bar = widgets.FloatProgress(value=0, min=0, max=100, description='Progress:', style={'description_width': '80px'})
        
        # Results
        self.result_tabs = widgets.Tab()
        self.pdf_output = widgets.Output()
        self.fig_gallery = widgets.Output()
        self.data_preview = widgets.Output()
        self.result_tabs.children = [self.pdf_output, self.fig_gallery, self.data_preview]
        self.result_tabs.set_title(0, '📄 PDF Report')
        self.result_tabs.set_title(1, '🖼️ Figure Gallery')
        self.result_tabs.set_title(2, '📊 Data Preview')
        
        # Layout
        left_panel = widgets.VBox([
            api_box, self.provider_dropdown, self.model_input, self.api_key_input,
            file_box, self.file_upload, self.file_preview,
            cond_box, self.flow_inputs, self.baseline_input, self.sampling_input,
            widgets.HTML('<br>'), self.generate_btn, self.progress_bar, self.progress_output
        ], layout=widgets.Layout(width='450px', padding='10px'))
        
        self.main_ui = widgets.HBox([left_panel, self.result_tabs], layout=widgets.Layout(flex='1'))
        display(widgets.VBox([self.header_box, self.main_ui]))
        
        # File preview observer
        self.file_upload.observe(self._on_file_change, names='value')

    def _on_file_change(self, change):
        with self.file_preview:
            clear_output()
            name, content = get_uploaded_file(self.file_upload)
            if name:
                print("✅ File loaded:", name)
                print("   Size: {:.1f} KB".format(len(content) / 1024 if content else 0))

    def _update_progress(self, val, msg=""):
        self.progress_bar.value = val
        with self.progress_output:
            clear_output()
            if msg: print(msg)

    def _on_generate(self, btn):
        try:
            self._update_progress(5, "⏳ Starting report generation...")
            
            # Update config
            self.cfg.ai_provider = self.provider_dropdown.value
            self.cfg.model_name = self.model_input.value
            self.cfg.api_key = self.api_key_input.value
            try:
                self.cfg.flow_conditions = [float(x.strip()) for x in self.flow_inputs.value.split(',') if x.strip()]
            except: pass
            self.cfg.baseline_flow = self.baseline_input.value
            self.cfg.sampling_rate = self.sampling_input.value
            
            self._update_progress(10, "⏳ Loading data...")
            # Load data
            name, content = get_uploaded_file(self.file_upload)
            self.loader = FlowLoopDataLoader(self.cfg, raw_bytes=content)
            self.data = self.loader.load()
            
            self._update_progress(30, "⏳ Running analysis...")
            # Analysis
            engine = AnalysisEngine(self.cfg, self.data)
            self.results = engine.run_full_analysis()
            
            self._update_progress(50, "⏳ Generating figures...")
            # Figures
            fig_gen = FigureGenerator(self.cfg, self.data, self.results)
            self.fig_paths = fig_gen.generate_all()
            
            self._update_progress(70, "⏳ Generating AI conclusions...")
            # AI
            ai_text = generate_ai_conclusions(
                self.cfg.ai_provider, self.cfg.api_key, self.cfg.model_name,
                self.data, self.results
            )
            
            self._update_progress(85, "⏳ Building PDF...")
            # PDF
            builder = PDFBuilder(self.cfg, self.data, self.results, self.fig_paths, ai_text, self.loader)
            self.pdf_path = builder.build()
            
            self._update_progress(100, "✅ Report generated successfully!")
            
            # Show results
            self._show_results()
            
        except Exception as e:
            self._update_progress(0, "❌ Error: {}".format(str(e)))
            import traceback
            traceback.print_exc()

    def _show_results(self):
        # PDF Tab
        with self.pdf_output:
            clear_output()
            if self.pdf_path and Path(self.pdf_path).exists():
                display(widgets.HTML("""
                    <div style="background: #d4edda; padding: 15px; border-radius: 8px; margin-bottom: 10px;">
                        <h4 style="color: #155724; margin: 0 0 5px 0;">✅ PDF Generated Successfully</h4>
                        <p style="margin: 0; color: #155724;">{path}</p>
                    </div>
                """.format(path=self.pdf_path)))
                with open(self.pdf_path, 'rb') as f:
                    pdf_data = f.read()
                b64 = base64.b64encode(pdf_data).decode()
                display(widgets.HTML("""
                    <iframe src="data:application/pdf;base64,{b64}" width="100%" height="600px" 
                            style="border: 1px solid #ccc; border-radius: 4px;"></iframe>
                    <br><br>
                    <a href="data:application/pdf;base64,{b64}" download="flowrate_effects_report.pdf"
                       style="background: #1f4e78; color: white; padding: 10px 20px; text-decoration: none; border-radius: 4px; display: inline-block;">
                        📥 Download PDF
                    </a>
                """.format(b64=b64)))
        
        # Figure Gallery
        with self.fig_gallery:
            clear_output()
            display(widgets.HTML('<h4 style="color: #1f4e78;">Generated Figures</h4>'))
            if self.fig_paths:
                for name, path in self.fig_paths.items():
                    with open(path, 'rb') as f:
                        img_data = f.read()
                    b64 = base64.b64encode(img_data).decode()
                    display(widgets.HTML("""
                        <div style="display: inline-block; margin: 10px; text-align: center;">
                            <img src="data:image/png;base64,{b64}" style="max-width: 350px; border: 1px solid #ddd; border-radius: 4px; cursor: pointer;"
                                 onclick="window.open(this.src, '_blank')">
                            <br><small style="color: #666;">{name} (click to enlarge)</small>
                        </div>
                    """.format(b64=b64, name=name)))
        
        # Data Preview
        with self.data_preview:
            clear_output()
            display(widgets.HTML('<h4 style="color: #1f4e78;">Data Summary</h4>'))
            if self.results:
                if 'step_up_summary' in self.results:
                    display(widgets.HTML('<h5>Step-Up Summary</h5>'))
                    display(self.results['step_up_summary'])
                if 'step_down_summary' in self.results:
                    display(widgets.HTML('<h5>Step-Down Summary</h5>'))
                    display(self.results['step_down_summary'])
                if 'ss_summary' in self.results:
                    display(widgets.HTML('<h5>Steady-State Summary</h5>'))
                    display(self.results['ss_summary'])
        
        # Switch to PDF tab
        self.result_tabs.selected_index = 0

# ══════════════════════════════════════════════════════════════════
#  MAIN ENTRY POINT
# ══════════════════════════════════════════════════════════════════

def main():
    print("=" * 60)
    print("  FLOW LOOP REPORT GENERATOR")
    print("  Professional Edition with AI Analysis")
    print("=" * 60)
    ui = FlowLoopReportUI()
    return ui

if __name__ == "__main__":
    ui = main()
else:
    ui = main()
```
